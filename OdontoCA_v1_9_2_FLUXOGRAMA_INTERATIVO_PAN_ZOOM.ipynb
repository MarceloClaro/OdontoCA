## Política de tratamento para melhorar desempenho sem inflar métricas

A melhoria de desempenho deve ser obtida por **curadoria e seleção de configuração
no conjunto de validação**, e não por uso repetido do conjunto de teste.

1. duplicatas exatas entre splits são removidas com prioridade `test > val > train`;
2. near-duplicates perceptuais são apenas sinalizadas;
3. classes são auditadas e, quando necessário, o treino pode ser reamostrado com
   limite de multiplicação;
4. panorâmicas utilizam resolução maior quando a lesão ocupa pequena fração da imagem;
5. transformações geométricas e fotométricas são conservadoras;
6. `flipud` é proibido;
7. panorâmicas não recebem jitter HSV;
8. tuning compara configurações somente em `val`;
9. o `test` é acessado apenas após seleção e treinamento final.

Isso diferencia **melhoria real de generalização** de simples otimização do conjunto
de avaliação.

````python
# Renderizar fluxograma interativo v1.9.2 — zoom, pan e navegação
from IPython.display import HTML, display
import html as _html

MERMAID_ODONTOCA_V192 = r"""
flowchart TB

INICIO([OdontoCA v1.9.2<br/>SDD 5.0 + TDD 5.0<br/>DADOS REAIS • ADAPTADORES UNIVERSAIS • QC • ANTI-LEAKAGE])

subgraph GOV["CAMADA 0 — GOVERNANÇA, SDD 5.0 E TDD 5.0"]
direction LR
G1["46 requisitos SDD"]
G2["58 testes TDD"]
G3["UNIT"]
G4["INTEGRATION"]
G5["ACCEPTANCE"]
G6["Rastreabilidade<br/>SDD → TDD"]
G7["Seeds • Proveniência • Hashes • LGPD"]
end

INICIO --> G1 --> G2
G2 --> G3
G2 --> G4
G2 --> G5
G3 --> G6
G4 --> G6
G5 --> G6
G6 --> G7

subgraph DADOS["CAMADA 1 — FONTES DE DADOS REAIS"]
direction LR
D1["Table_S1.xlsx<br/>Single-tooth ECC"]
D2["Qiita 14341<br/>BIOM + mapping"]
D3["Zenodo 14827784<br/>fotografia intraoral"]
D4["Children Caries Detection<br/>panorâmica pediátrica"]
D5["Children's Dental Panoramic<br/>Zhang et al."]
D6["Dental Anatomy YOLOv8"]
D7["SegmentAnyTooth<br/>FDI opcional"]
D8["REGRA<br/>sem métricas científicas sintéticas"]
end

G7 --> D1
G7 --> D2
G7 --> D3
G7 --> D4
G7 --> D5
G7 --> D6
G7 -.-> D7
D1 -.-> D8
D2 -.-> D8
D3 -.-> D8
D4 -.-> D8
D5 -.-> D8
D6 -.-> D8

subgraph CLIN["CAMADA 2 — CLÍNICA LONGITUDINAL"]
direction TB
C1["Ler Table_S1<br/>schema externo preservado"]
C2["Contrato de colunas<br/>sum_dmfs • sum_ns_dt • sum_s_dt"]
C3["Excluir T5161"]
C4["Organizar criança × dente × tempo"]
C5["Reconstruir t → t+1"]
C6{"Δ timepoint = 1?"}
C7["Coorte onset<br/>estado inicial H"]
C8["Target<br/>0=H→H • 1=H→C"]
C9["Auditoria de reprodução<br/>2504→2284→1388→1160→997/84"]
C10["Features clínico-espaciais"]
C11["Autômato celular"]
C12["Nested Group CV por criança"]
C13["Modelo clínico-espacial"]
end

D1 --> C1 --> C2 --> C3 --> C4 --> C5 --> C6
C6 -- Sim --> C7 --> C8 --> C9 --> C10 --> C11 --> C12 --> C13
C6 -- Não --> CB["Excluir transição"]

subgraph MIC["CAMADA 3 — MICROBIOMA"]
direction TB
M1["Inventariar Qiita/BIOM"]
M2["Pareamento determinístico<br/>SampleID ↔ BIOM"]
M3{"Pareamento único?"}
M4["Matriz ASV real"]
M5["Auditar library size"]
M6{"Biblioteca > 0?"}
M7["Alinhamento target × SampleID"]
M8["Outer Group CV"]
M9["Inner Group CV"]
M10["Filtro ASV<br/>FIT apenas treino"]
M11["Pseudocontagem + CLR"]
M12["StandardScaler"]
M13["PCA"]
M14["Tuning nested"]
M15["Modelo microbioma"]
M16["Modelo combinado<br/>clínico + microbioma"]
end

D2 --> M1 --> M2 --> M3
M3 -- Sim --> M4 --> M5 --> M6
M3 -- Não --> MB["BLOCKED<br/>pareamento ambíguo"]
M6 -- Sim --> M7 --> M8 --> M9 --> M10 --> M11 --> M12 --> M13 --> M14 --> M15
M6 -- Não --> MZ["BLOCKED<br/>biblioteca zero"]
C13 --> M16
M15 --> M16

subgraph INVENT["CAMADA 4 — INVENTÁRIO E DETECÇÃO AUTOMÁTICA DE ANOTAÇÕES"]
direction TB
I1["Extrair ZIP real"]
I2["Inventariar arquivos<br/>imagens • TXT • JSON • XML • CSV • YAML"]
I3{"Formato reconhecido?"}
I4["YOLO YAML oficial"]
I5["TXT YOLO por stem global"]
I6["LabelMe JSON"]
I7["COCO JSON"]
I8["Pascal VOC XML"]
I9["CSV bounding boxes"]
I10["Máscaras → bbox YOLO"]
I11["Converter para YOLO canônico"]
I12["Registrar proveniência da conversão"]
I13["Registrar falhas de parsing"]
I14["Imagens sem ground truth<br/>→ inference_only"]
end

D3 --> I1
D4 --> I1
D5 --> I1
D6 --> I1
I1 --> I2 --> I3
I3 --> I4
I3 --> I5
I3 --> I6
I3 --> I7
I3 --> I8
I3 --> I9
I3 --> I10
I4 --> I11
I5 --> I11
I6 --> I11
I7 --> I11
I8 --> I11
I9 --> I11
I10 --> I11
I11 --> I12 --> I13
I11 --> I14

subgraph SPLIT["CAMADA 5 — POLÍTICA DE SPLIT SEM LEAKAGE"]
direction TB
S1{"Split oficial do autor?"}
S2["Preservar train/val/test"]
S3{"patient_id confiável?"}
S4["Group split por paciente"]
S5["Criar apenas train + val"]
S6["NÃO criar test artificial"]
S7["Registrar nível de independência"]
end

I11 --> S1
S1 -- Sim --> S2 --> S7
S1 -- Não --> S3
S3 -- Sim --> S4 --> S7
S3 -- Não --> S5 --> S6 --> S7

subgraph QC["CAMADA 6 — CONTROLE DE QUALIDADE VISUAL"]
direction TB
Q1["Resolução<br/>largura × altura"]
Q2["Intensidade média"]
Q3["Contraste"]
Q4["Variância do Laplaciano"]
Q5["Blur suspeito"]
Q6["Subexposição"]
Q7["Superexposição"]
Q8["Baixo contraste"]
Q9["Sem exclusão automática por padrão"]
end

S7 --> Q1 --> Q2 --> Q3 --> Q4 --> Q5
Q4 --> Q6
Q4 --> Q7
Q4 --> Q8
Q5 --> Q9
Q6 --> Q9
Q7 --> Q9
Q8 --> Q9

subgraph CURA["CAMADA 7 — CURADORIA ANTI-LEAKAGE"]
direction TB
V1["SHA-256 por imagem"]
V2["Remover duplicatas exatas<br/>prioridade test > val > train"]
V3["dHash + BK-tree<br/>near-duplicates reportados"]
V4["Auditar classes e áreas bbox"]
V5{"Desbalanceamento severo?"}
V6["Reamostragem limitada<br/>somente train"]
V7["val/test inalterados"]
V8["Dataset tratado"]
end

Q9 --> V1 --> V2 --> V3 --> V4 --> V5
V5 -- Sim --> V6 --> V7 --> V8
V5 -- Não --> V7 --> V8

subgraph DOM["CAMADA 8 — DOMÍNIOS VISUAIS SEPARADOS"]
direction LR
A1["Fotografia intraoral<br/>Zenodo<br/>imgsz 768"]
A2["Panorâmica pediátrica — cárie<br/>imgsz 960"]
A3["Panorâmica pediátrica — doenças<br/>imgsz 960"]
A4["Anatomia dentária<br/>imgsz 768"]
end

V8 --> A1
V8 --> A2
V8 --> A3
V8 --> A4

subgraph TREINO["CAMADA 9 — TUNING E TREINO FINAL"]
direction TB
T1["Configuração-base segura"]
T2["Configuração tratada"]
T3["Treinos curtos"]
T4["Comparar SOMENTE em val"]
T5{"Selecionar melhor configuração"}
T6["Treino final"]
T7["Early stopping"]
T8["Salvar best.pt"]
T9["SHA-256 best.pt"]
T10{"Test legítimo disponível?"}
T11["Avaliar uma vez em test"]
T12["Sem test legítimo:<br/>avaliar val e declarar limitação"]
T13["Precision • Recall<br/>mAP50 • mAP50-95"]
end

A1 --> T1
A1 --> T2
A2 --> T1
A2 --> T2
A3 --> T1
A3 --> T2
A4 --> T1
A4 --> T2
T1 --> T3
T2 --> T3
T3 --> T4 --> T5 --> T6 --> T7 --> T8 --> T9 --> T10
T10 -- Sim --> T11 --> T13
T10 -- Não --> T12 --> T13

subgraph FDI["CAMADA 10 — FDI E ODONTOGRAMA"]
direction TB
F1["SegmentAnyTooth"]
F2["Máscara por dente"]
F3["Numeração FDI"]
F4["Detecção de cárie"]
F5["IoU/overlap lesão × dente"]
F6{"Associação não ambígua?"}
F7["Probabilidade por dente"]
F8["Odontograma automático"]
F9["Revisão profissional"]
end

D7 --> F1 --> F2 --> F3
T13 --> F4
F3 --> F5
F4 --> F5
F5 --> F6
F6 -- Sim --> F7 --> F8
F6 -- Não --> F9

subgraph FUSAO["CAMADA 11 — FUSÃO MULTIMODAL"]
direction TB
U1["Predição clínica"]
U2["Predição microbioma"]
U3["Predição visual por dente"]
U4{"Mesma coorte?"}
U5{"Mesmo patient_id?"}
U6{"Mesmo visit_id?"}
U7{"Mesmo tooth_fdi?"}
U8["Fusão autorizada"]
U9["Meta-modelo"]
UB["BLOCKED<br/>cross-cohort/key mismatch"]
end

C13 --> U1
M15 --> U2
F7 --> U3
U1 --> U4
U2 --> U4
U3 --> U4
U4 -- Não --> UB
U4 -- Sim --> U5
U5 -- Não --> UB
U5 -- Sim --> U6
U6 -- Não --> UB
U6 -- Sim --> U7
U7 -- Não --> UB
U7 -- Sim --> U8 --> U9

subgraph VAL["CAMADA 12 — VALIDAÇÃO CIENTÍFICA"]
direction LR
R1["OOF"]
R2["PR-AUC"]
R3["ROC-AUC"]
R4["Brier"]
R5["Log Loss"]
R6["Bootstrap por criança"]
R7["Δ métricas"]
R8["Below-chance audit"]
R9["1-AUC<br/>diagnóstico apenas"]
end

C13 --> R1
M15 --> R1
M16 --> R1
R1 --> R2
R1 --> R3
R1 --> R4
R1 --> R5
R1 --> R6 --> R7
R1 --> R8 --> R9

subgraph TDDRUN["CAMADA 13 — EXECUÇÃO TDD 5.0"]
direction TB
Z1["Executar 58 testes"]
Z2["UNIT"]
Z3["INTEGRATION"]
Z4["ACCEPTANCE"]
Z5["PASS / SKIP / FAIL"]
Z6["Rastreabilidade SDD → TDD"]
Z7{"Existe FAIL?"}
Z8["Cobertura SDD = 100%"]
ZB["PIPELINE BLOQUEADO"]
end

R9 --> Z1
T13 --> Z1
F8 --> Z1
Z1 --> Z2
Z1 --> Z3
Z1 --> Z4
Z2 --> Z5
Z3 --> Z5
Z4 --> Z5
Z5 --> Z6 --> Z7
Z7 -- Sim --> ZB
Z7 -- Não --> Z8

subgraph GATES["CAMADA 14 — CRITÉRIOS DE ACEITAÇÃO"]
direction TB
AC1["Contagens clínicas reproduzidas"]
AC2["Leakage paciente = 0"]
AC3["Qiita alinhado"]
AC4["Nested preprocessing"]
AC5["Formato visual reconhecido"]
AC6["Ground truth suficiente"]
AC7["Inference-only separado"]
AC8["QC visual disponível"]
AC9["Duplicatas cross-split = 0"]
AC10["Tuning somente em val"]
AC11["Rotas legadas removidas"]
AC12["best.pt + SHA-256"]
AC13["Métricas finitas"]
AC14{"Todos aprovados?"}
AP["APPROVED<br/>PIPELINE DE PESQUISA"]
BL["BLOCKED<br/>corrigir e reexecutar"]
NOTE["APPROVED ≠ validação clínica"]
end

Z8 --> AC1 --> AC2 --> AC3 --> AC4 --> AC5 --> AC6 --> AC7 --> AC8 --> AC9 --> AC10 --> AC11 --> AC12 --> AC13 --> AC14
AC14 -- Sim --> AP --> NOTE
AC14 -- Não --> BL

subgraph OUT["CAMADA 15 — SAÍDAS REPRODUZÍVEIS"]
direction LR
O1["sdd_5_requisitos.csv"]
O2["tdd_5_especificacoes.csv"]
O3["tdd_5_resultados.csv"]
O4["rastreabilidade_sdd_tdd.csv"]
O5["inventarios de datasets"]
O6["falhas_parsing_v1_9.csv"]
O7["QC visual CSVs"]
O8["predicoes OOF"]
O9["metricas modelos"]
O10["tuning_visual_validacao.csv"]
O11["best.pt + hashes"]
O12["manifesto"]
O13["relatório de validação"]
O14["FHIR RiskAssessment<br/>preliminary"]
O15["ZIP final"]
end

AP --> O1
AP --> O2
AP --> O3
AP --> O4
AP --> O5
AP --> O6
AP --> O7
AP --> O8
AP --> O9
AP --> O10
AP --> O11
AP --> O12
AP --> O13
AP --> O14
AP --> O15

O15 --> FIM([RESULTADOS PARA ARTIGO<br/>somente após execução real aprovada])

classDef gov fill:#E8EAF6,stroke:#3949AB,stroke-width:2px,color:#111;
classDef dados fill:#E8F5E9,stroke:#2E7D32,stroke-width:2px,color:#111;
classDef clin fill:#E3F2FD,stroke:#1565C0,stroke-width:2px,color:#111;
classDef mic fill:#F3E5F5,stroke:#6A1B9A,stroke-width:2px,color:#111;
classDef visual fill:#FCE4EC,stroke:#C2185B,stroke-width:2px,color:#111;
classDef treino fill:#FFF3E0,stroke:#EF6C00,stroke-width:2px,color:#111;
classDef fusao fill:#E0F2F1,stroke:#00796B,stroke-width:2px,color:#111;
classDef val fill:#FFF8E1,stroke:#F9A825,stroke-width:2px,color:#111;
classDef tdd fill:#EDE7F6,stroke:#512DA8,stroke-width:2px,color:#111;
classDef out fill:#E0F7FA,stroke:#00838F,stroke-width:2px,color:#111;
classDef block fill:#FFEBEE,stroke:#C62828,stroke-width:3px,color:#B71C1C;
classDef ok fill:#E8F5E9,stroke:#1B5E20,stroke-width:3px,color:#1B5E20;

class G1,G2,G3,G4,G5,G6,G7 gov;
class D1,D2,D3,D4,D5,D6,D7,D8 dados;
class C1,C2,C3,C4,C5,C6,C7,C8,C9,C10,C11,C12,C13 clin;
class M1,M2,M3,M4,M5,M6,M7,M8,M9,M10,M11,M12,M13,M14,M15,M16 mic;
class I1,I2,I3,I4,I5,I6,I7,I8,I9,I10,I11,I12,I13,I14,Q1,Q2,Q3,Q4,Q5,Q6,Q7,Q8,Q9,V1,V2,V3,V4,V5,V6,V7,V8,A1,A2,A3,A4,F1,F2,F3,F4,F5,F6,F7,F8,F9 visual;
class T1,T2,T3,T4,T5,T6,T7,T8,T9,T10,T11,T12,T13 treino;
class U1,U2,U3,U4,U5,U6,U7,U8,U9 fusao;
class R1,R2,R3,R4,R5,R6,R7,R8,R9 val;
class Z1,Z2,Z3,Z4,Z5,Z6,Z7,Z8 tdd;
class O1,O2,O3,O4,O5,O6,O7,O8,O9,O10,O11,O12,O13,O14,O15 out;
class MB,MZ,CB,UB,ZB,BL block;
class AP,NOTE,FIM ok;
"""

# A interface abaixo é independente do pipeline científico.
# Se o navegador bloquear a CDN do Mermaid, apenas o fluxograma deixa de renderizar;
# as demais células podem continuar normalmente.
_DOCUMENTO_FLUXOGRAMA_INTERATIVO = r"""
<!DOCTYPE html>
<html>
<head>
<meta charset="utf-8">
<style>
  * { box-sizing: border-box; }

  html, body {
    margin: 0;
    padding: 0;
    background: #f8fafc;
    font-family: Arial, Helvetica, sans-serif;
    color: #172033;
  }

  #app {
    display: flex;
    flex-direction: column;
    width: 100%;
    height: 100%;
    min-height: 900px;
    background: #f8fafc;
  }

  #barra {
    position: sticky;
    top: 0;
    z-index: 20;
    display: flex;
    flex-wrap: wrap;
    align-items: center;
    gap: 7px;
    padding: 9px 10px;
    background: rgba(255,255,255,.97);
    border-bottom: 1px solid #d8dee9;
    box-shadow: 0 1px 4px rgba(15,23,42,.08);
  }

  .grupo {
    display: flex;
    align-items: center;
    gap: 5px;
    padding-right: 8px;
    margin-right: 2px;
    border-right: 1px solid #e5e7eb;
  }

  .grupo:last-child { border-right: 0; }

  button, select {
    height: 34px;
    border: 1px solid #cbd5e1;
    border-radius: 7px;
    background: white;
    color: #172033;
    font-size: 13px;
  }

  button {
    min-width: 38px;
    padding: 0 10px;
    cursor: pointer;
    font-weight: 600;
  }

  button:hover {
    background: #eef4ff;
    border-color: #93b4e8;
  }

  button:active { transform: translateY(1px); }

  button:focus-visible, select:focus-visible {
    outline: 3px solid rgba(37,99,235,.24);
    outline-offset: 1px;
  }

  select {
    min-width: 240px;
    padding: 0 8px;
    cursor: pointer;
  }

  #zoomValor {
    min-width: 58px;
    text-align: center;
    font-size: 12px;
    font-weight: 700;
    color: #334155;
  }

  #status {
    margin-left: auto;
    min-width: 190px;
    text-align: right;
    font-size: 12px;
    color: #64748b;
  }

  #viewport {
    position: relative;
    flex: 1 1 auto;
    width: 100%;
    height: 820px;
    overflow: hidden;
    background:
      linear-gradient(#e8edf4 1px, transparent 1px),
      linear-gradient(90deg, #e8edf4 1px, transparent 1px),
      #ffffff;
    background-size: 28px 28px;
    cursor: grab;
    touch-action: none;
    user-select: none;
  }

  #viewport.arrastando { cursor: grabbing; }

  #stage {
    position: absolute;
    left: 0;
    top: 0;
    transform-origin: 0 0;
    will-change: transform;
  }

  #stage svg {
    display: block;
    max-width: none !important;
    height: auto;
    background: white;
  }

  #ajuda {
    position: absolute;
    right: 12px;
    bottom: 12px;
    z-index: 10;
    max-width: 330px;
    padding: 8px 10px;
    border: 1px solid #d8dee9;
    border-radius: 8px;
    background: rgba(255,255,255,.94);
    box-shadow: 0 2px 7px rgba(15,23,42,.08);
    color: #475569;
    font-size: 11px;
    line-height: 1.45;
    pointer-events: none;
  }

  #minimapa {
    position: absolute;
    left: 12px;
    bottom: 12px;
    z-index: 10;
    width: 180px;
    height: 112px;
    overflow: hidden;
    border: 1px solid #cbd5e1;
    border-radius: 8px;
    background: rgba(255,255,255,.94);
    box-shadow: 0 2px 7px rgba(15,23,42,.08);
  }

  #miniConteudo {
    position: absolute;
    left: 0;
    top: 0;
    transform-origin: 0 0;
    pointer-events: none;
  }

  #miniConteudo svg {
    display: block;
    max-width: none !important;
  }

  #miniJanela {
    position: absolute;
    border: 2px solid #2563eb;
    background: rgba(37,99,235,.09);
    pointer-events: none;
  }

  #erro {
    display: none;
    margin: 16px;
    padding: 12px;
    border: 1px solid #ef9a9a;
    border-radius: 8px;
    background: #fff5f5;
    color: #b42318;
    white-space: pre-wrap;
  }

  @media (max-width: 720px) {
    #barra { gap: 5px; }
    select { min-width: 170px; max-width: 220px; }
    #status { width: 100%; text-align: left; margin-left: 0; }
    #viewport { height: 700px; }
    #minimapa { width: 145px; height: 92px; }
    #ajuda { max-width: 210px; }
  }
</style>
</head>
<body>
<div id="app">

  <div id="barra" role="toolbar" aria-label="Controles do fluxograma">
    <div class="grupo">
      <button id="menos" type="button" title="Reduzir">−</button>
      <span id="zoomValor">100%</span>
      <button id="mais" type="button" title="Ampliar">+</button>
      <button id="cem" type="button" title="Zoom de 100%">100%</button>
      <button id="ajustar" type="button" title="Ajustar todo o fluxograma">Ajustar</button>
    </div>

    <div class="grupo" aria-label="Navegação">
      <button id="esquerda" type="button" title="Mover para a esquerda">←</button>
      <button id="cima" type="button" title="Mover para cima">↑</button>
      <button id="baixo" type="button" title="Mover para baixo">↓</button>
      <button id="direita" type="button" title="Mover para a direita">→</button>
      <button id="centro" type="button" title="Centralizar">Centro</button>
    </div>

    <div class="grupo">
      <label for="irPara" style="font-size:12px;font-weight:700;">Ir para:</label>
      <select id="irPara" aria-label="Ir para camada do fluxograma">
        <option value="">Selecione uma camada...</option>
        <option value="CAMADA 0">0 — Governança / SDD / TDD</option>
        <option value="CAMADA 1">1 — Fontes de dados reais</option>
        <option value="CAMADA 2">2 — Clínica longitudinal</option>
        <option value="CAMADA 3">3 — Microbioma</option>
        <option value="CAMADA 4">4 — Inventário e anotações</option>
        <option value="CAMADA 5">5 — Política de split</option>
        <option value="CAMADA 6">6 — Controle de qualidade</option>
        <option value="CAMADA 7">7 — Curadoria anti-leakage</option>
        <option value="CAMADA 8">8 — Domínios visuais</option>
        <option value="CAMADA 9">9 — Tuning e treino final</option>
        <option value="CAMADA 10">10 — FDI e odontograma</option>
        <option value="CAMADA 11">11 — Fusão multimodal</option>
        <option value="CAMADA 12">12 — Validação científica</option>
        <option value="CAMADA 13">13 — Execução TDD 5.0</option>
        <option value="CAMADA 14">14 — Critérios de aceitação</option>
        <option value="CAMADA 15">15 — Saídas reproduzíveis</option>
      </select>
    </div>

    <div id="status" aria-live="polite">Carregando fluxograma...</div>
  </div>

  <div id="viewport" tabindex="0" aria-label="Fluxograma interativo. Arraste para navegar e use a roda do mouse para zoom.">
    <div id="stage">
      <div id="mermaidPrincipal" class="mermaid">__MERMAID__</div>
    </div>

    <div id="minimapa" aria-hidden="true">
      <div id="miniConteudo"></div>
      <div id="miniJanela"></div>
    </div>

    <div id="ajuda">
      <b>Navegação:</b> arraste com mouse/toque para mover.<br>
      <b>Zoom:</b> roda do mouse, botões +/− ou teclas +/−.<br>
      <b>Atalhos:</b> setas movem; 0 = ajustar; 1 = 100%.<br>
      Use <b>Ir para</b> para saltar entre as 16 camadas.
    </div>
  </div>

  <div id="erro"></div>
</div>

<script type="module">
(async () => {
  const viewport = document.getElementById("viewport");
  const stage = document.getElementById("stage");
  const status = document.getElementById("status");
  const erro = document.getElementById("erro");
  const zoomValor = document.getElementById("zoomValor");
  const mini = document.getElementById("minimapa");
  const miniConteudo = document.getElementById("miniConteudo");
  const miniJanela = document.getElementById("miniJanela");

  let escala = 1;
  let x = 0;
  let y = 0;
  let svg = null;
  let diagramaW = 1;
  let diagramaH = 1;

  const LIM_MIN = 0.12;
  const LIM_MAX = 4.5;

  function clamp(v, min, max) {
    return Math.min(max, Math.max(min, v));
  }

  function aplicarTransformacao() {
    stage.style.transform = `translate(${x}px, ${y}px) scale(${escala})`;
    zoomValor.textContent = `${Math.round(escala * 100)}%`;
    atualizarMinimapa();
  }

  function dimensoesDiagrama() {
    if (!svg) return;
    const vb = svg.viewBox && svg.viewBox.baseVal;
    if (vb && vb.width > 0 && vb.height > 0) {
      diagramaW = vb.width;
      diagramaH = vb.height;
      svg.style.width = `${diagramaW}px`;
      svg.style.height = `${diagramaH}px`;
    } else {
      const box = svg.getBBox();
      diagramaW = Math.max(1, box.width);
      diagramaH = Math.max(1, box.height);
    }
  }

  function ajustar() {
    if (!svg) return;
    dimensoesDiagrama();
    const margem = 34;
    const vw = Math.max(100, viewport.clientWidth - margem * 2);
    const vh = Math.max(100, viewport.clientHeight - margem * 2);
    escala = clamp(Math.min(vw / diagramaW, vh / diagramaH), LIM_MIN, 1.25);
    x = (viewport.clientWidth - diagramaW * escala) / 2;
    y = (viewport.clientHeight - diagramaH * escala) / 2;
    aplicarTransformacao();
    status.textContent = "Fluxograma ajustado à janela.";
  }

  function centralizar() {
    if (!svg) return;
    x = (viewport.clientWidth - diagramaW * escala) / 2;
    y = (viewport.clientHeight - diagramaH * escala) / 2;
    aplicarTransformacao();
  }

  function zoomPara(novaEscala, px = viewport.clientWidth/2, py = viewport.clientHeight/2) {
    if (!svg) return;
    novaEscala = clamp(novaEscala, LIM_MIN, LIM_MAX);
    const mundoX = (px - x) / escala;
    const mundoY = (py - y) / escala;
    escala = novaEscala;
    x = px - mundoX * escala;
    y = py - mundoY * escala;
    aplicarTransformacao();
  }

  function mover(dx, dy) {
    x += dx;
    y += dy;
    aplicarTransformacao();
  }

  function zoom100() {
    escala = 1;
    centralizar();
    status.textContent = "Zoom em 100%.";
  }

  function encontrarTexto(termo) {
    if (!svg) return null;
    const termoUpper = termo.toUpperCase();
    const candidatos = [...svg.querySelectorAll("text, foreignObject")];
    return candidatos.find(el => (el.textContent || "").toUpperCase().includes(termoUpper)) || null;
  }

  function irParaCamada(termo) {
    if (!termo || !svg) return;
    const alvo = encontrarTexto(termo);
    if (!alvo) {
      status.textContent = `Camada não localizada: ${termo}`;
      return;
    }

    let box;
    try {
      box = alvo.getBBox();
    } catch {
      const r = alvo.getBoundingClientRect();
      const vr = viewport.getBoundingClientRect();
      const cxTela = r.left + r.width/2 - vr.left;
      const cyTela = r.top + r.height/2 - vr.top;
      x += viewport.clientWidth/2 - cxTela;
      y += viewport.clientHeight/2 - cyTela;
      aplicarTransformacao();
      return;
    }

    const cx = box.x + box.width / 2;
    const cy = box.y + box.height / 2;
    const alvoEscala = Math.max(escala, 0.75);
    escala = clamp(alvoEscala, LIM_MIN, LIM_MAX);
    x = viewport.clientWidth/2 - cx * escala;
    y = viewport.clientHeight/2 - cy * escala;
    aplicarTransformacao();
    status.textContent = `Navegação: ${termo}.`;
  }

  function criarMinimapa() {
    if (!svg) return;
    miniConteudo.innerHTML = "";
    const clone = svg.cloneNode(true);
    clone.removeAttribute("id");
    clone.querySelectorAll("[id]").forEach(el => el.removeAttribute("id"));

    const mw = mini.clientWidth;
    const mh = mini.clientHeight;
    const s = Math.min(mw / diagramaW, mh / diagramaH) * 0.95;

    clone.style.width = `${diagramaW}px`;
    clone.style.height = `${diagramaH}px`;
    miniConteudo.style.transform = `translate(${(mw-diagramaW*s)/2}px, ${(mh-diagramaH*s)/2}px) scale(${s})`;
    miniConteudo.dataset.scale = String(s);
    miniConteudo.dataset.ox = String((mw-diagramaW*s)/2);
    miniConteudo.dataset.oy = String((mh-diagramaH*s)/2);
    miniConteudo.appendChild(clone);
    atualizarMinimapa();
  }

  function atualizarMinimapa() {
    if (!svg || !miniConteudo.dataset.scale) return;
    const ms = Number(miniConteudo.dataset.scale);
    const ox = Number(miniConteudo.dataset.ox);
    const oy = Number(miniConteudo.dataset.oy);

    const mundoEsq = (0 - x) / escala;
    const mundoTopo = (0 - y) / escala;
    const mundoLarg = viewport.clientWidth / escala;
    const mundoAlt = viewport.clientHeight / escala;

    miniJanela.style.left = `${ox + mundoEsq * ms}px`;
    miniJanela.style.top = `${oy + mundoTopo * ms}px`;
    miniJanela.style.width = `${Math.max(8, mundoLarg * ms)}px`;
    miniJanela.style.height = `${Math.max(8, mundoAlt * ms)}px`;
  }

  // Controles
  document.getElementById("mais").addEventListener("click", () => zoomPara(escala * 1.22));
  document.getElementById("menos").addEventListener("click", () => zoomPara(escala / 1.22));
  document.getElementById("cem").addEventListener("click", zoom100);
  document.getElementById("ajustar").addEventListener("click", ajustar);
  document.getElementById("centro").addEventListener("click", centralizar);
  document.getElementById("esquerda").addEventListener("click", () => mover(100, 0));
  document.getElementById("direita").addEventListener("click", () => mover(-100, 0));
  document.getElementById("cima").addEventListener("click", () => mover(0, 100));
  document.getElementById("baixo").addEventListener("click", () => mover(0, -100));
  document.getElementById("irPara").addEventListener("change", (e) => irParaCamada(e.target.value));

  // Zoom pela roda do mouse centrado no ponteiro
  viewport.addEventListener("wheel", (e) => {
    e.preventDefault();
    const r = viewport.getBoundingClientRect();
    const px = e.clientX - r.left;
    const py = e.clientY - r.top;
    const fator = Math.exp(-e.deltaY * 0.0012);
    zoomPara(escala * fator, px, py);
  }, { passive:false });

  // Pan por Pointer Events (mouse, caneta e toque)
  let arrastando = false;
  let ultimoX = 0;
  let ultimoY = 0;

  viewport.addEventListener("pointerdown", (e) => {
    if (e.target.closest && e.target.closest("#barra")) return;
    arrastando = true;
    ultimoX = e.clientX;
    ultimoY = e.clientY;
    viewport.classList.add("arrastando");
    viewport.setPointerCapture(e.pointerId);
  });

  viewport.addEventListener("pointermove", (e) => {
    if (!arrastando) return;
    const dx = e.clientX - ultimoX;
    const dy = e.clientY - ultimoY;
    ultimoX = e.clientX;
    ultimoY = e.clientY;
    mover(dx, dy);
  });

  function terminarArrasto(e) {
    arrastando = false;
    viewport.classList.remove("arrastando");
    try { viewport.releasePointerCapture(e.pointerId); } catch {}
  }

  viewport.addEventListener("pointerup", terminarArrasto);
  viewport.addEventListener("pointercancel", terminarArrasto);

  // Teclado quando o viewport estiver focado
  viewport.addEventListener("keydown", (e) => {
    if (["INPUT","SELECT","TEXTAREA"].includes(e.target.tagName)) return;
    const passo = e.shiftKey ? 180 : 80;
    if (e.key === "+" || e.key === "=") { e.preventDefault(); zoomPara(escala * 1.2); }
    else if (e.key === "-") { e.preventDefault(); zoomPara(escala / 1.2); }
    else if (e.key === "0") { e.preventDefault(); ajustar(); }
    else if (e.key === "1") { e.preventDefault(); zoom100(); }
    else if (e.key === "ArrowLeft") { e.preventDefault(); mover(passo, 0); }
    else if (e.key === "ArrowRight") { e.preventDefault(); mover(-passo, 0); }
    else if (e.key === "ArrowUp") { e.preventDefault(); mover(0, passo); }
    else if (e.key === "ArrowDown") { e.preventDefault(); mover(0, -passo); }
  });

  // Mini mapa clicável
  mini.addEventListener("pointerdown", (e) => {
    if (!svg || !miniConteudo.dataset.scale) return;
    const r = mini.getBoundingClientRect();
    const ms = Number(miniConteudo.dataset.scale);
    const ox = Number(miniConteudo.dataset.ox);
    const oy = Number(miniConteudo.dataset.oy);
    const mundoX = (e.clientX-r.left-ox) / ms;
    const mundoY = (e.clientY-r.top-oy) / ms;
    x = viewport.clientWidth/2 - mundoX*escala;
    y = viewport.clientHeight/2 - mundoY*escala;
    aplicarTransformacao();
  });

  window.addEventListener("resize", () => {
    criarMinimapa();
  });

  try {
    const mermaidModulo = await import("https://cdn.jsdelivr.net/npm/mermaid@11/dist/mermaid.esm.min.mjs");
    const mermaid = mermaidModulo.default;

    mermaid.initialize({
      startOnLoad: false,
      securityLevel: "loose",
      theme: "base",
      flowchart: {
        htmlLabels: true,
        curve: "basis",
        nodeSpacing: 30,
        rankSpacing: 42,
        useMaxWidth: false
      },
      themeVariables: {
        fontFamily: "Arial, Helvetica, sans-serif",
        fontSize: "13px"
      }
    });

    await mermaid.run({ querySelector:"#mermaidPrincipal" });

    svg = document.querySelector("#mermaidPrincipal svg");
    if (!svg) throw new Error("SVG Mermaid não foi criado.");

    svg.removeAttribute("width");
    svg.removeAttribute("height");
    svg.style.maxWidth = "none";
    dimensoesDiagrama();

    // Impede o SVG de capturar gestos de arrasto.
    svg.style.pointerEvents = "none";

    criarMinimapa();
    ajustar();

    status.textContent = "Pronto — arraste para navegar; roda do mouse ou +/− para zoom.";
  } catch (e) {
    erro.style.display = "block";
    erro.textContent =
      "Não foi possível renderizar o fluxograma interativo.\\n" +
      (e && e.message ? e.message : String(e)) +
      "\\n\\nO pipeline científico pode continuar normalmente.";
    status.textContent = "Falha apenas na visualização.";
  }
})();
</script>
</body>
</html>
"""

try:
    _documento = _DOCUMENTO_FLUXOGRAMA_INTERATIVO.replace(
        "__MERMAID__",
        _html.escape(MERMAID_ODONTOCA_V192)
    )

    _iframe = (
        '<iframe '
        'style="width:100%;height:1040px;border:1px solid #cbd5e1;'
        'border-radius:12px;background:#f8fafc;" '
        'srcdoc="' + _html.escape(_documento, quote=True) + '">'
        '</iframe>'
    )
    display(HTML(_iframe))
    print("Fluxograma interativo v1.9.2 carregado: zoom, pan, minimapa e navegação por camada.")
except Exception as erro_fluxograma:
    print("AVISO: falha apenas na interface do fluxograma.")
    print(type(erro_fluxograma).__name__, ":", erro_fluxograma)
````

````python
# 4. Aquisição robusta e verificação de integridade
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

def criar_sessao_http():
    s = requests.Session()
    s.mount('https://', HTTPAdapter(max_retries=Retry(total=5, backoff_factor=1.0, status_forcelist=[429, 500, 502, 503, 504])))
    return s

def baixar_arquivo_em_blocos(url, destino, timeout=1200):
    destino = Path(destino)
    temporario = destino.with_suffix(destino.suffix + '.part')
    with criar_sessao_http().get(url, stream=True, timeout=timeout) as r:
        r.raise_for_status()
        with open(temporario, 'wb') as f:
            for bloco in r.iter_content(1024 * 1024):
                if bloco:
                    f.write(bloco)
    temporario.replace(destino)
    return destino

def calcular_sha256(caminho, bloco=1024 * 1024):
    objeto_hash = hashlib.sha256()
    with open(caminho, 'rb') as f:
        for b in iter(lambda: f.read(bloco), b''):
            objeto_hash.update(b)
    return objeto_hash.hexdigest()

def calcular_md5(caminho, bloco=1024 * 1024):
    objeto_hash = hashlib.md5()
    with open(caminho, 'rb') as f:
        for b in iter(lambda: f.read(bloco), b''):
            objeto_hash.update(b)
    return objeto_hash.hexdigest()
````

````python
# 5. Ingestão do dataset clínico longitudinal real
# IMPORTANTE:
# - Os nomes das colunas do arquivo-fonte permanecem EXATAMENTE como publicados.
# - Somente após a leitura os dados recebem nomes internos em português.
# - Isto evita quebrar a compatibilidade com Table_S1.xlsx.

COLUNAS_FONTE_OBRIGATORIAS = [
    "SampleID", "HostGroup", "Status_Host", "Status_Tooth",
    "dmfs_Host", "dmfs_Tooth", "Month", "Timepoint", "Position",
    "NicheICM", "NicheUL", "NicheLR", "HostID",
    "Future_Status_Tooth", "NicheAP", "Position_Tooth",
    "X", "Y", "Z", "spatial_dist_weighted_mean_dmfs",
    "sum_dmfs", "sum_ns_dt", "sum_s_dt", "Time_to_decay",
]

# Compatibilidade defensiva com versões antigas do próprio notebook.
# Se um arquivo já tiver sido renomeado indevidamente para português,
# normalizamos de volta para o contrato oficial da fonte.
ALIAS_COLUNAS_FONTE = {
    "soma_dmfs": "sum_dmfs",
    "soma_ns_dt": "sum_ns_dt",
    "soma_s_dt": "sum_s_dt",
}

def carregar_tabela_s1_real():
    destino_alvo = PASTAS["brutos"] / "Table_S1.xlsx"
    arquivo_local = (
        Path(CONFIG.CAMINHO_LOCAL_TABELA_S1)
        if CONFIG.CAMINHO_LOCAL_TABELA_S1
        else None
    )

    if arquivo_local and arquivo_local.exists():
        shutil.copy2(arquivo_local, destino_alvo)

    if not destino_alvo.exists():
        print("Baixando Table_S1.xlsx real...")
        baixar_arquivo_em_blocos(
            CONFIG.URL_TABELA_S1,
            destino_alvo,
            timeout=300,
        )

    arquivo_excel = pd.ExcelFile(destino_alvo)
    if "all_metadata" not in arquivo_excel.sheet_names:
        raise RuntimeError(
            "A planilha 'all_metadata' não foi encontrada. "
            f"Planilhas disponíveis: {arquivo_excel.sheet_names}"
        )

    metadados = pd.read_excel(
        destino_alvo,
        sheet_name="all_metadata",
    )
    metadados.columns = [str(coluna).strip() for coluna in metadados.columns]

    # Normalização apenas de aliases conhecidos.
    renomeacoes = {
        antiga: oficial
        for antiga, oficial in ALIAS_COLUNAS_FONTE.items()
        if antiga in metadados.columns and oficial not in metadados.columns
    }
    if renomeacoes:
        print("Normalizando aliases de colunas da fonte:", renomeacoes)
        metadados = metadados.rename(columns=renomeacoes)

    ausentes = [
        coluna
        for coluna in COLUNAS_FONTE_OBRIGATORIAS
        if coluna not in metadados.columns
    ]

    if ausentes:
        print("\nColunas detectadas no arquivo:")
        print(sorted(metadados.columns.tolist()))
        raise RuntimeError(
            "Campos obrigatórios ausentes no Table_S1.xlsx: "
            f"{ausentes}. "
            "Os nomes externos não devem ser traduzidos no arquivo-fonte."
        )

    print("Table_S1 SHA-256:", calcular_sha256(destino_alvo))
    print("Linhas:", len(metadados))
    print("Crianças:", metadados["HostID"].nunique())
    print("Contrato de colunas externas: APROVADO")

    return metadados, destino_alvo

METADADOS, CAMINHO_TABELA_S1 = carregar_tabela_s1_real()
````

````python
# 6. Curadoria da coorte e representação por dente
ARCADA_SUPERIOR = ['T55', 'T54', 'T53', 'T52', 'T51', 'T61', 'T62', 'T63', 'T64', 'T65']
ARCADA_INFERIOR = ['T85', 'T84', 'T83', 'T82', 'T81', 'T71', 'T72', 'T73', 'T74', 'T75']
ARCADAS = {'upper': ARCADA_SUPERIOR, 'lower': ARCADA_INFERIOR}

# Mapeamento anatÃ´mico de simetria bilateral (antÃ­meros homÃ³logos contralaterais)
MAPA_ANTIMEROS = {
    'T51': 'T61', 'T61': 'T51',
    'T52': 'T62', 'T62': 'T52',
    'T53': 'T63', 'T63': 'T53',
    'T54': 'T64', 'T64': 'T54',
    'T55': 'T65', 'T65': 'T55',
    'T81': 'T71', 'T71': 'T81',
    'T82': 'T72', 'T72': 'T82',
    'T83': 'T73', 'T73': 'T83',
    'T84': 'T74', 'T74': 'T84',
    'T85': 'T75', 'T75': 'T85',
}

# Ãndice biolÃ³gico de vulnerabilidade salivar (1=protegido por saliva/baixa, 2=fÃ³ssula oclusal/mÃ©dia, 3=mamadeira noturna/alta)
VULNERABILIDADE_SALIVAR = {
    'T51': 3, 'T52': 3, 'T61': 3, 'T62': 3,
    'T54': 2, 'T55': 2, 'T64': 2, 'T65': 2,
    'T84': 2, 'T85': 2, 'T74': 2, 'T75': 2,
    'T53': 1, 'T63': 1, 'T83': 1, 'T73': 1,
    'T81': 1, 'T82': 1, 'T71': 1, 'T72': 1,
}


def construir_mapa_dentario(metadados):
    linhas = []
    for arcada, sequencia in ARCADAS.items():
        for indices, dente in enumerate(sequencia):
            fonte = metadados.loc[metadados['Position'].eq(dente)]
            if len(fonte) == 0:
                raise RuntimeError(f'Tooth position ausente: {dente}')
            r = fonte.iloc[0]
            linhas.append({'posicao_dente': dente, 'dente_fdi': int(dente[1:]), 'indice_tipo_dente': int(str(r['Position_Tooth'])[1:]), 'arcada': arcada, 'indice_arcada': indices, 'nicho_icm': r['NicheICM'], 'nicho_ul': r['NicheUL'], 'nicho_lr': r['NicheLR'], 'nicho_ap': r['NicheAP'], 'x': float(r['X']), 'y': float(r['Y']), 'z': float(r['Z']), 'dente_vizinho_anterior': sequencia[indices - 1] if indices > 0 else np.nan, 'dente_vizinho_seguinte': sequencia[indices + 1] if indices < len(sequencia) - 1 else np.nan})
    return pd.DataFrame(linhas)

def limpar_registros_dentes_individuais(metadados):
    tooth_map = construir_mapa_dentario(metadados)
    registros_compostos_local = metadados.loc[metadados['Position'].eq('T5161')].copy()
    registros_individuais = metadados.loc[metadados['Position'].isin(tooth_map['posicao_dente'])].copy()
    registros_individuais = registros_individuais.merge(tooth_map, left_on='Position', right_on='posicao_dente', how='left', validate='many_to_one')
    dados_limpos = pd.DataFrame({'id_amostra': registros_individuais['SampleID'].astype(str), 'id_crianca': registros_individuais['HostID'].astype(str), 'ponto_temporal': registros_individuais['Timepoint'].astype(str), 'indice_temporal': registros_individuais['Timepoint'].astype(str).str.extract('(\\d+)')[0].astype(int), 'idade_meses': pd.to_numeric(registros_individuais['Month'], errors='coerce'), 'posicao_dente': registros_individuais['posicao_dente'], 'dente_fdi': registros_individuais['dente_fdi'], 'indice_tipo_dente': registros_individuais['indice_tipo_dente'], 'arcada': registros_individuais['arcada'], 'indice_arcada': registros_individuais['indice_arcada'], 'nicho_icm': registros_individuais['NicheICM'], 'nicho_ul': registros_individuais['NicheUL'], 'nicho_lr': registros_individuais['NicheLR'], 'nicho_ap': registros_individuais['NicheAP'], 'x': pd.to_numeric(registros_individuais['X'], errors='coerce'), 'y': pd.to_numeric(registros_individuais['Y'], errors='coerce'), 'z': pd.to_numeric(registros_individuais['Z'], errors='coerce'), 'estado_dente': registros_individuais['Status_Tooth'].astype(str), 'dmfs_dente': pd.to_numeric(registros_individuais['dmfs_Tooth'], errors='coerce'), 'estado_hospedeiro': registros_individuais['Status_Host'].astype(str), 'dmfs_hospedeiro': pd.to_numeric(registros_individuais['dmfs_Host'], errors='coerce'), 'dmfs_espacial_ponderado': pd.to_numeric(registros_individuais['spatial_dist_weighted_mean_dmfs'], errors='coerce'), 'soma_dmfs': pd.to_numeric(registros_individuais['sum_dmfs'], errors='coerce'), 'soma_ns_dt': pd.to_numeric(registros_individuais['sum_ns_dt'], errors='coerce'), 'soma_s_dt': pd.to_numeric(registros_individuais['sum_s_dt'], errors='coerce'), 'dente_vizinho_anterior': registros_individuais['dente_vizinho_anterior'], 'dente_vizinho_seguinte': registros_individuais['dente_vizinho_seguinte'], 'grupo_hospedeiro_fonte': registros_individuais['HostGroup'], 'estado_futuro_fonte': registros_individuais['Future_Status_Tooth'], 'tempo_ate_carie_fonte': pd.to_numeric(registros_individuais['Time_to_decay'], errors='coerce')})
    if dados_limpos.duplicated(['id_crianca', 'posicao_dente', 'ponto_temporal']).any():
        raise RuntimeError('Duplicidade criança-dente-timepoint.')
    return (tooth_map, registros_compostos_local, dados_limpos)
MAPA_DENTARIO, REGISTROS_COMPOSTOS, DADOS_DENTES_LIMPOS = limpar_registros_dentes_individuais(METADADOS)
print('T5161:', len(REGISTROS_COMPOSTOS), 'Single-tooth:', len(DADOS_DENTES_LIMPOS))
````

````python
# 7. Reconstrução longitudinal das transições t → t+1
def construir_transicoes(dados_limpos):
    consulta_vizinhos = dados_limpos.set_index(['id_crianca', 'ponto_temporal', 'posicao_dente'])[['estado_dente', 'dmfs_dente']]
    s = dados_limpos.sort_values(['id_crianca', 'posicao_dente', 'indice_temporal', 'id_amostra']).copy()
    g = s.groupby(['id_crianca', 'posicao_dente'], sort=False)
    for fonte, destino_alvo in [('id_amostra', 'id_amostra_t1'), ('ponto_temporal', 'ponto_temporal_t1'), ('indice_temporal', 'indice_temporal_t1'), ('idade_meses', 'idade_meses_t1'), ('estado_dente', 'estado_dente_t1'), ('dmfs_dente', 'dmfs_dente_t1'), ('estado_hospedeiro', 'estado_hospedeiro_t1'), ('dmfs_hospedeiro', 'dmfs_hospedeiro_t1')]:
        s[destino_alvo] = g[fonte].shift(-1)
    observacoes = s.loc[s['id_amostra_t1'].notna()].copy()
    observacoes['delta_pontos_temporais'] = observacoes['indice_temporal_t1'] - observacoes['indice_temporal']
    observacoes['delta_meses'] = observacoes['idade_meses_t1'] - observacoes['idade_meses']
    observacoes['evento_progressao'] = ((observacoes['estado_dente'] == 'H') & (observacoes['estado_dente_t1'] == 'C')).astype(int)
    observacoes['classe_transicao'] = observacoes['estado_dente'] + '->' + observacoes['estado_dente_t1'].astype(str)

    def obter_vizinho(row, coluna, campo):
        dente = row[coluna]
        if pd.isna(dente):
            return np.nan
        try:
            return consulta_vizinhos.loc[(row['id_crianca'], row['ponto_temporal'], dente), campo]
        except KeyError:
            return np.nan
    observacoes['estado_vizinho_anterior_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_vizinho_anterior', 'estado_dente'), axis=1)
    observacoes['dmfs_vizinho_anterior_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_vizinho_anterior', 'dmfs_dente'), axis=1)
    observacoes['estado_vizinho_seguinte_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_vizinho_seguinte', 'estado_dente'), axis=1)
    observacoes['dmfs_vizinho_seguinte_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_vizinho_seguinte', 'dmfs_dente'), axis=1)
    observacoes['n_vizinhos_observados_t'] = observacoes[['estado_vizinho_anterior_t', 'estado_vizinho_seguinte_t']].notna().sum(axis=1)
    # Features topolÃ³gicas por dente: simetria bilateral (antÃ­mero) e gradiente salivar
    observacoes['dente_antimero'] = observacoes['posicao_dente'].map(MAPA_ANTIMEROS)
    observacoes['estado_antimero_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_antimero', 'estado_dente'), axis=1)
    observacoes['dmfs_antimero_t'] = observacoes.apply(lambda r: obter_vizinho(r, 'dente_antimero', 'dmfs_dente'), axis=1)
    observacoes['antimero_com_carie_t'] = (observacoes['estado_antimero_t'] == 'C').astype(int)
    observacoes['indice_vulnerabilidade_salivar'] = observacoes['posicao_dente'].map(VULNERABILIDADE_SALIVAR).fillna(2).astype(float)
    consecutivas = observacoes.loc[observacoes['delta_pontos_temporais'].eq(1)].copy()
    coorte_onset = consecutivas.loc[consecutivas['estado_dente'].eq('H')].copy()
    coorte_onset['y'] = (coorte_onset['estado_dente_t1'] == 'C').astype(int)
    coorte_onset['n_vizinhos_com_carie'] = (coorte_onset['estado_vizinho_anterior_t'] == 'C').astype(int) + (coorte_onset['estado_vizinho_seguinte_t'] == 'C').astype(int)
    return (observacoes, consecutivas, coorte_onset)
TRANSICOES, TRANSICOES_CONSECUTIVAS, COORTE_ONSET = construir_transicoes(DADOS_DENTES_LIMPOS)
````

````python
# 9. Modelo clínico-espacial real — validação OOF por criança
PREDITORES_PROIBIDOS = {'HostGroup', 'Future_Status_Tooth', 'Time_to_decay', 'estado_dente_t1', 'y'}
COLUNAS_CATEGORICAS = ['nicho_icm', 'nicho_ul', 'nicho_lr', 'nicho_ap', 'estado_hospedeiro']
COLUNAS_NUMERICAS = ['idade_meses', 'dmfs_hospedeiro', 'dmfs_espacial_ponderado', 'soma_dmfs', 'soma_ns_dt', 'soma_s_dt', 'dmfs_vizinho_anterior_t', 'dmfs_vizinho_seguinte_t', 'n_vizinhos_observados_t', 'n_vizinhos_com_carie', 'dmfs_antimero_t', 'antimero_com_carie_t', 'indice_vulnerabilidade_salivar']
COORTE_ONSET['estado_hospedeiro'] = COORTE_ONSET['estado_hospedeiro']
assert PREDITORES_PROIBIDOS.isdisjoint(set(COLUNAS_CATEGORICAS + COLUNAS_NUMERICAS))

def criar_codificador_one_hot_denso():
    try:
        return OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown='ignore', sparse=False)

def criar_pipeline_clinico(seed):
    preprocessador = ColumnTransformer([('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', criar_codificador_one_hot_denso())]), COLUNAS_CATEGORICAS), ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), COLUNAS_NUMERICAS)], remainder='drop', sparse_threshold=0)
    return Pipeline([('prep', preprocessador), ('clf', LogisticRegression(penalty='l2', solver='liblinear', max_iter=5000, random_state=seed))])

def calcular_numero_particoes_seguro(alvo, grupos, solicitado):
    temporario = pd.DataFrame({'y': np.asarray(alvo, int), 'g': np.asarray(grupos, str)})
    por_grupo = temporario.groupby('g')['y'].max()
    contagens = por_grupo.value_counts()
    if len(contagens) < 2:
        raise RuntimeError('Classes de grupos insuficientes para CV.')
    n = min(int(solicitado), int(contagens.min()))
    if n < 2:
        raise RuntimeError('Menos de dois folds independentes possíveis.')
    return n

def obter_probabilidade_classe_positiva(estimador, matriz_x, classe_positiva=1):
    classes = np.asarray(estimador.classes_)
    indices = np.flatnonzero(classes == classe_positiva)
    if len(indices) != 1:
        raise RuntimeError(f'Classe positiva não identificada: {classes}')
    return estimador.predict_proba(matriz_x)[:, int(indices[0])]

def validar_oof_aninhado_clinico(dados):
    alvo = dados['y'].astype(int).to_numpy()
    grupos = dados['id_crianca'].astype(str).to_numpy()
    n_particoes_externas = calcular_numero_particoes_seguro(alvo, grupos, CONFIG.PARTICOES_EXTERNAS)
    validacao_externa = StratifiedGroupKFold(n_splits=n_particoes_externas, shuffle=True, random_state=CONFIG.SEMENTE)
    predicao = np.full(len(dados), np.nan)
    particao = np.full(len(dados), -1)
    auditoria = []
    for f, (indices_treino, indices_teste) in enumerate(validacao_externa.split(dados, alvo, grupos)):
        sobreposicao = set(grupos[indices_treino]).intersection(set(grupos[indices_teste]))
        if sobreposicao:
            raise RuntimeError(f'Patient leakage no fold {f}: {sobreposicao}')
        n_particoes_internas = calcular_numero_particoes_seguro(alvo[indices_treino], grupos[indices_treino], CONFIG.PARTICOES_INTERNAS)
        validacao_interna = StratifiedGroupKFold(n_splits=n_particoes_internas, shuffle=True, random_state=CONFIG.SEMENTE + 100 + f)
        busca_grade = GridSearchCV(criar_pipeline_clinico(CONFIG.SEMENTE + f), {'clf__C': [0.01, 0.1, 1.0, 10.0]}, scoring='average_precision', cv=validacao_interna, n_jobs=-1, refit=True, error_score='raise')
        busca_grade.fit(dados.iloc[indices_treino][COLUNAS_CATEGORICAS + COLUNAS_NUMERICAS], alvo[indices_treino], groups=grupos[indices_treino])
        predicao[indices_teste] = obter_probabilidade_classe_positiva(busca_grade.best_estimator_, dados.iloc[indices_teste][COLUNAS_CATEGORICAS + COLUNAS_NUMERICAS], 1)
        particao[indices_teste] = f
        auditoria.append({'particao': f, 'n_treino': len(indices_treino), 'n_teste': len(indices_teste), 'eventos_teste': int(alvo[indices_teste].sum()), 'sobreposicao_pacientes': len(sobreposicao), 'melhores_parametros': json.dumps(busca_grade.best_params_)})
    if not np.isfinite(predicao).all():
        raise RuntimeError('Predições OOF incompletas.')
    metricas = {'modelo': 'Regressão logística clínico-espacial aninhada', 'n': len(alvo), 'eventos': int(alvo.sum()), 'prevalencia': float(alvo.mean()), 'PR_AUC': float(average_precision_score(alvo, predicao)), 'ROC_AUC': float(roc_auc_score(alvo, predicao)), 'Brier': float(brier_score_loss(alvo, predicao)), 'LogLoss': float(log_loss(alvo, np.clip(predicao, 1e-06, 1 - 1e-06)))}
    return (metricas, predicao, particao, pd.DataFrame(auditoria))
METRICAS_CLINICAS, PREDICOES_OOF_CLINICAS, PARTICAO_CLINICA, AUDITORIA_PARTICOES_CLINICAS = validar_oof_aninhado_clinico(COORTE_ONSET)
display(pd.DataFrame([METRICAS_CLINICAS]))
display(AUDITORIA_PARTICOES_CLINICAS)
````

# 10. Microbioma real — Qiita Study 14341

Esta etapa baixa o pacote público do Qiita e tenta parear cada amostra ao
`SampleID` clínico do conjunto de onset. O pipeline é **fail-fast** para
pareamentos ambíguos, IDs duplicados e bibliotecas zeradas.

````python
# 10.1 Dependência biom-format + bridge determinístico Qiita
if importlib.util.find_spec('biom') is None:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'biom-format'], check=True)
from biom import load_table
QIITA_DOWNLOAD = f'https://qiita.ucsd.edu/public_download/?data=biom&study_id={CONFIG.ID_ESTUDO_QIITA}'

def normalizar_id_clinico(id_bruto, ids_alvo, mapeamento_explicito=None):
    mapeamento_explicito = mapeamento_explicito or {}
    ids_alvo = set(map(str, ids_alvo))
    mapa_maiusculo = {x.upper(): x for x in ids_alvo}
    if id_bruto is None:
        return (None, 'missing')
    s = str(id_bruto).strip()
    if not s:
        return (None, 'empty')
    if s in ids_alvo:
        return (s, 'exact')
    texto_maiusculo = s.upper()
    if texto_maiusculo in mapa_maiusculo:
        return (mapa_maiusculo[texto_maiusculo], 'case_normalized')
    segmentos = [t for t in re.split('[._:/|\\s-]+', texto_maiusculo) if t]
    correspondencias = list(dict.fromkeys((mapa_maiusculo[t] for t in segmentos if t in mapa_maiusculo)))
    if len(correspondencias) == 1:
        return (correspondencias[0], 'namespace_token')
    if len(correspondencias) > 1:
        return (None, 'ambiguous_namespace_token')
    ids_embutidos = re.findall('(?i)A\\d{4,6}', s)
    correspondencias = list(dict.fromkeys((mapa_maiusculo[x.upper()] for x in ids_embutidos if x.upper() in mapa_maiusculo)))
    if len(correspondencias) == 1:
        return (correspondencias[0], 'embedded_A_id')
    if len(correspondencias) > 1:
        return (None, 'ambiguous_embedded_A_id')
    if s in mapeamento_explicito and mapeamento_explicito[s] in ids_alvo:
        return (mapeamento_explicito[s], 'qiita_mapping_file')
    return (None, 'unmatched')

def inferir_id_artefato(caminho):
    caminho = Path(caminho)
    for parte in reversed(caminho.parts):
        if re.fullmatch('\\d{5,7}', str(parte)):
            return str(parte)
    correspondencia = re.search('(?<!\\d)(\\d{5,7})(?!\\d)', caminho.name)
    return correspondencia.group(1) if correspondencia else ''

def descobrir_mapeamento_qiita_explicito(pasta_qiita, todos_ids_amostras_brutas, ids_alvo):
    """
    Procura arquivos tabulares de mapping/metadata dentro do pacote Qiita.
    Só aceita raw->clinical quando um raw ID aponta para exatamente um
    SampleID clínico.
    """
    candidatos = []
    tabular = [p for p in pasta_qiita.rglob('*') if p.is_file() and p.suffix.lower() in {'.txt', '.tsv', '.csv'}]
    for caminho_mapeamento in tabular:
        try:
            tabela_mapeamento = pd.read_csv(caminho_mapeamento, sep=None, engine='python', dtype=str, nrows=300000)
        except Exception:
            continue
        if tabela_mapeamento.empty or len(tabela_mapeamento.columns) < 2:
            continue
        colunas_id_bruto = []
        for c in tabela_mapeamento.columns:
            vals = set(tabela_mapeamento[c].dropna().astype(str).str.strip())
            if vals & todos_ids_amostras_brutas:
                colunas_id_bruto.append(c)
        for coluna_id_bruto in colunas_id_bruto:
            for coluna_clinica in tabela_mapeamento.columns:
                if coluna_clinica == coluna_id_bruto:
                    continue
                temporario = tabela_mapeamento[[coluna_id_bruto, coluna_clinica]].dropna().copy()
                temporario.columns = ['raw', 'clin']
                temporario['raw'] = temporario['raw'].astype(str).str.strip()
                temporario['cid'] = temporario['clin'].map(lambda x: normalizar_id_clinico(x, ids_alvo)[0])
                temporario = temporario[temporario['raw'].isin(todos_ids_amostras_brutas) & temporario['cid'].notna()]
                if len(temporario):
                    candidatos.extend(temporario[['raw', 'cid']].itertuples(index=False, name=None))
    if not candidatos:
        return {}
    tabela_mapeamento_unico = pd.DataFrame(candidatos, columns=['raw', 'cid']).drop_duplicates()
    n_unicos = tabela_mapeamento_unico.groupby('raw')['cid'].nunique()
    ids_validos = set(n_unicos[n_unicos == 1].index)
    mapeamento_explicito_local = tabela_mapeamento_unico[tabela_mapeamento_unico['raw'].isin(ids_validos)].drop_duplicates('raw').set_index('raw')['cid'].to_dict()
    return mapeamento_explicito_local

def carregar_contagens_qiita_reais(ids_alvo):
    zip_qiita_local = PASTAS['brutos'] / f'qiita_{CONFIG.ID_ESTUDO_QIITA}.zip'
    pasta_qiita = PASTAS['brutos'] / f'qiita_{CONFIG.ID_ESTUDO_QIITA}'
    pasta_qiita.mkdir(exist_ok=True)
    arquivo_local = Path(CONFIG.CAMINHO_LOCAL_ZIP_QIITA) if CONFIG.CAMINHO_LOCAL_ZIP_QIITA else None
    if arquivo_local and arquivo_local.exists():
        shutil.copy2(arquivo_local, zip_qiita_local)
    if not zip_qiita_local.exists():
        print('Baixando Qiita Study 14341...')
        baixar_arquivo_em_blocos(QIITA_DOWNLOAD, zip_qiita_local, timeout=1200)
    if not zipfile.is_zipfile(zip_qiita_local):
        raise RuntimeError('Download Qiita não é ZIP válido.')
    if not any(pasta_qiita.iterdir()):
        with zipfile.ZipFile(zip_qiita_local) as quadro_modelagem_local:
            quadro_modelagem_local.extractall(pasta_qiita)
    arquivos_biom = sorted(pasta_qiita.rglob('*.biom'))
    if not arquivos_biom:
        raise RuntimeError('Nenhum BIOM encontrado no pacote Qiita.')
    ids_alvo = set(map(str, ids_alvo))
    tabelas = {}
    todos_ids_brutos = set()
    for caminho_biom in arquivos_biom:
        try:
            tabela_biom = load_table(str(caminho_biom))
        except Exception:
            continue
        tabelas[str(caminho_biom)] = tabela_biom
        todos_ids_brutos.update(map(str, tabela_biom.ids(axis='sample')))
    if not tabelas:
        raise RuntimeError('Nenhum BIOM pôde ser aberto.')
    mapeamento_explicito = descobrir_mapeamento_qiita_explicito(pasta_qiita, todos_ids_brutos, ids_alvo)
    inventario = []
    for caminho, tabela_biom in tabelas.items():
        resolved = {}
        metodos = {}
        for id_bruto in map(str, tabela_biom.ids(axis='sample')):
            id_clinico, metodo = normalizar_id_clinico(id_bruto, ids_alvo, mapeamento_explicito)
            if id_clinico:
                resolved.setdefault(id_clinico, []).append(id_bruto)
                metodos[id_bruto] = metodo
        n_colisoes = sum((len(set(v)) > 1 for v in resolved.values()))
        tamanhos_biblioteca = np.asarray(tabela_biom.sum(axis='sample')).astype(float)
        id_artefato_temp = inferir_id_artefato(caminho)
        inventario.append({'id_artefato': id_artefato_temp, 'path': caminho, 'nome_arquivo': Path(caminho).name, 'n_amostras': len(tabela_biom.ids(axis='sample')), 'n_variaveis': len(tabela_biom.ids(axis='observation')), 'sobreposicao_alvo': len(resolved), 'n_colisoes': n_colisoes, 'mediana_tamanho_biblioteca': float(np.median(tamanhos_biblioteca)) if len(tamanhos_biblioteca) else np.nan, 'eh_all_biom': Path(caminho).name.lower() == 'all.biom', 'artefato_preferido': id_artefato_temp == str(CONFIG.ID_ARTEFATO_QIITA_PREFERIDO)})
    inventario_local = pd.DataFrame(inventario)
    if len(inventario_local) == 0:
        raise RuntimeError('Inventário Qiita vazio.')
    validos = inventario_local[inventario_local['sobreposicao_alvo'].fillna(0).gt(0) & inventario_local['n_colisoes'].fillna(0).eq(0)].copy()
    if len(validos) == 0:
        display(inventario_local)
        raise RuntimeError('Nenhum BIOM com pareamento determinístico sem colisões.')
    validos['preferencia'] = 10000 * validos['artefato_preferido'].astype(int) + 1000 * validos['eh_all_biom'].astype(int) + validos['sobreposicao_alvo'].fillna(0)
    validos = validos.sort_values(['preferencia', 'sobreposicao_alvo', 'mediana_tamanho_biblioteca'], ascending=[False, False, False])
    selecionados = []
    ids_cobertos = set()
    for _, r in validos.iterrows():
        tabela_biom = tabelas[r['path']]
        resolved = {}
        for id_bruto in map(str, tabela_biom.ids(axis='sample')):
            id_clinico, metodo = normalizar_id_clinico(id_bruto, ids_alvo, mapeamento_explicito)
            if id_clinico:
                resolved[id_clinico] = (id_bruto, metodo)
        novos_ids = set(resolved) - ids_cobertos
        if novos_ids:
            linha_selecionada = r.to_dict()
            linha_selecionada['new_target_samples_added'] = len(novos_ids)
            selecionados.append(linha_selecionada)
            ids_cobertos |= novos_ids
    selected_df = pd.DataFrame(selecionados)
    atribuicoes = []
    for _, r in selected_df.iterrows():
        tabela_biom = tabelas[r['path']]
        amostras = list(map(str, tabela_biom.ids(axis='sample')))
        somas = np.asarray(tabela_biom.sum(axis='sample')).astype(float)
        mapa_somas = dict(zip(amostras, somas))
        for id_bruto in amostras:
            id_clinico, metodo = normalizar_id_clinico(id_bruto, ids_alvo, mapeamento_explicito)
            if id_clinico:
                atribuicoes.append({'id_amostra_clinica': id_clinico, 'id_amostra_biom_bruto': id_bruto, 'id_artefato': r['id_artefato'], 'path': r['path'], 'metodo_pareamento': metodo, 'preferencia': r['preferencia'], 'tamanho_biblioteca': mapa_somas.get(id_bruto, np.nan)})
    atribuicoes_finais = pd.DataFrame(atribuicoes)
    atribuicoes_finais = atribuicoes_finais.sort_values(['id_amostra_clinica', 'preferencia', 'tamanho_biblioteca'], ascending=[True, False, False]).drop_duplicates('id_amostra_clinica', keep='first')
    if atribuicoes_finais['id_amostra_clinica'].duplicated().any():
        raise RuntimeError('SampleID clínico recebeu mais de uma representação final.')
    linhas = {}
    fonte = []
    for _, a in atribuicoes_finais.iterrows():
        tabela_biom = tabelas[a['path']]
        id_bruto = str(a['id_amostra_biom_bruto'])
        id_clinico = str(a['id_amostra_clinica'])
        vetor = np.asarray(tabela_biom.data(id_bruto, axis='sample', dense=True)).astype(float)
        variaveis = list(map(str, tabela_biom.ids(axis='observation')))
        linhas[id_clinico] = {f: float(v) for f, v in zip(variaveis, vetor) if v > 0}
        fonte.append({'id_amostra': id_clinico, 'id_bruto': id_bruto, 'id_artefato': a['id_artefato'], 'metodo_pareamento': a['metodo_pareamento'], 'tamanho_biblioteca': float(vetor.sum()), 'n_variaveis_nao_zero': int((vetor > 0).sum())})
    contagens = pd.DataFrame.from_dict(linhas, orient='index').fillna(0.0).sort_index()
    contagens.index.name = 'id_amostra'
    if not contagens.index.is_unique:
        raise RuntimeError('Índice microbioma final não único.')
    if not set(contagens.index).issubset(ids_alvo):
        raise RuntimeError('Matriz contém SampleIDs fora da coorte alvo.')
    if (contagens.sum(axis=1) <= 0).any():
        raise RuntimeError('Biblioteca microbiológica zerada.')
    inventario_local.to_csv(PASTAS['microbioma'] / 'qiita_artifact_inventory.csv', index=False)
    selected_df.to_csv(PASTAS['microbioma'] / 'qiita_selected_artifacts.csv', index=False)
    atribuicoes_finais.to_csv(PASTAS['microbioma'] / 'atribuicoes_qiita_finais.csv', index=False)
    return (contagens, inventario_local, selected_df, atribuicoes_finais, pd.DataFrame(fonte), zip_qiita_local)
IDS_ALVO = set(COORTE_ONSET['id_amostra'].astype(str))
CONTAGENS_MICROBIOMA, INVENTARIO_QIITA, ARTEFATOS_QIITA_SELECIONADOS, ATRIBUICOES_QIITA, FONTE_MICROBIOMA, ZIP_QIITA = carregar_contagens_qiita_reais(IDS_ALVO)
print('Microbiome samples matched:', len(CONTAGENS_MICROBIOMA))
display(ARTEFATOS_QIITA_SELECIONADOS)
````

````python
# 12. Pré-processamento microbiológico completamente aninhado
class SeletorCLRMicrobioma(BaseEstimator, TransformerMixin):

    def __init__(self, prevalencia_minima=0.05, total_minimo=20.0, pseudocontagem=0.5, minimo_variaveis=5):
        self.prevalencia_minima = prevalencia_minima
        self.total_minimo = total_minimo
        self.pseudocontagem = pseudocontagem
        self.minimo_variaveis = minimo_variaveis

    def fit(self, matriz_x, alvo=None):
        matriz = matriz_x.to_numpy(float) if isinstance(matriz_x, pd.DataFrame) else np.asarray(matriz_x, float)
        self.nomes_variaveis_entrada_ = np.asarray(matriz_x.columns if isinstance(matriz_x, pd.DataFrame) else [f'f{i}' for i in range(matriz.shape[1])], dtype=object)
        prevalencia_por_variavel = (matriz > 0).mean(axis=0)
        total = matriz.sum(axis=0)
        indices_manter = np.flatnonzero((prevalencia_por_variavel >= self.prevalencia_minima) & (total >= self.total_minimo))
        if len(indices_manter) < self.minimo_variaveis:
            indices_manter = np.argsort(-prevalencia_por_variavel)[:min(self.minimo_variaveis, matriz.shape[1])]
        self.indices_mantidos_ = np.asarray(indices_manter, dtype=int)
        self.variaveis_mantidas_ = self.nomes_variaveis_entrada_[self.indices_mantidos_]
        return self

    def transform(self, matriz_x):
        matriz = matriz_x.to_numpy(float) if isinstance(matriz_x, pd.DataFrame) else np.asarray(matriz_x, float)
        matriz = matriz[:, self.indices_mantidos_] + float(self.pseudocontagem)
        log_contagens = np.log(matriz)
        return log_contagens - log_contagens.mean(axis=1, keepdims=True)

def criar_pipeline_microbioma(seed):
    return Pipeline([('clr', SeletorCLRMicrobioma()), ('scale', StandardScaler()), ('clf', LogisticRegression(penalty='l2', solver='liblinear', max_iter=5000, random_state=seed))])

def criar_pipeline_combinado(colunas_asv, seed):
    pipeline_micro = Pipeline([('clr', SeletorCLRMicrobioma()), ('scale', StandardScaler()), ('pca', PCA(n_components=0.95, svd_solver='full'))])
    preprocessador = ColumnTransformer([('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', criar_codificador_one_hot_denso())]), COLUNAS_CATEGORICAS), ('num', Pipeline([('imp', SimpleImputer(strategy='median')), ('sc', StandardScaler())]), COLUNAS_NUMERICAS), ('microbioma', pipeline_micro, colunas_asv)], remainder='drop', sparse_threshold=0)
    return Pipeline([('prep', preprocessador), ('clf', LogisticRegression(penalty='l2', solver='liblinear', max_iter=5000, random_state=seed))])
````

````python
# 13. Comparação OOF real — clínico vs microbioma vs combinado
def montar_quadro_variaveis_reais(dados_modelo, contagens):
    quadro_modelagem_local = dados_modelo.reset_index(drop=True).copy()
    matriz_contagens_local = contagens.reindex(quadro_modelagem_local['id_amostra'].astype(str)).copy()
    matriz_contagens_local.columns = [f'ASV__{c}' for c in matriz_contagens_local.columns.astype(str)]
    matriz_contagens_local = matriz_contagens_local.reset_index(drop=True)
    return (pd.concat([quadro_modelagem_local, matriz_contagens_local], axis=1), list(matriz_contagens_local.columns))

def validar_oof_aninhado_tres_modelos(dados_modelo, contagens):
    quadro_modelagem_local, colunas_asv = montar_quadro_variaveis_reais(dados_modelo, contagens)
    alvo = quadro_modelagem_local['y'].astype(int).to_numpy()
    grupos = quadro_modelagem_local['id_crianca'].astype(str).to_numpy()
    n_particoes_externas = calcular_numero_particoes_seguro(alvo, grupos, CONFIG.PARTICOES_EXTERNAS)
    validacao_externa = StratifiedGroupKFold(n_splits=n_particoes_externas, shuffle=True, random_state=CONFIG.SEMENTE)
    predicoes = {k: np.full(len(quadro_modelagem_local), np.nan) for k in ['clinico', 'microbioma', 'combinado']}
    id_particao = np.full(len(quadro_modelagem_local), -1)
    detalhes = []
    for f, (indices_treino, indices_teste) in enumerate(validacao_externa.split(quadro_modelagem_local, alvo, grupos)):
        sobreposicao = set(grupos[indices_treino]).intersection(set(grupos[indices_teste]))
        if sobreposicao:
            raise RuntimeError(f'Leakage de pacientes fold {f}: {sobreposicao}')
        n_particoes_internas = calcular_numero_particoes_seguro(alvo[indices_treino], grupos[indices_treino], CONFIG.PARTICOES_INTERNAS)
        validacao_interna = StratifiedGroupKFold(n_splits=n_particoes_internas, shuffle=True, random_state=CONFIG.SEMENTE + 100 + f)
        modelos = {'clinico': (criar_pipeline_clinico(CONFIG.SEMENTE + f), {'clf__C': [0.01, 0.1, 1, 10]}, COLUNAS_CATEGORICAS + COLUNAS_NUMERICAS), 'microbioma': (criar_pipeline_microbioma(CONFIG.SEMENTE + 1000 + f), {'clf__C': [0.001, 0.01, 0.1, 1]}, colunas_asv), 'combinado': (criar_pipeline_combinado(colunas_asv, CONFIG.SEMENTE + 2000 + f), {'clf__C': [0.01, 0.1, 1, 10]}, COLUNAS_CATEGORICAS + COLUNAS_NUMERICAS + colunas_asv)}
        row = {'particao': f, 'n_treino': len(indices_treino), 'n_teste': len(indices_teste), 'eventos_teste': int(alvo[indices_teste].sum()), 'sobreposicao_pacientes': 0}
        for nome, (pipeline_local, grade_parametros, colunas) in modelos.items():
            busca_grade = GridSearchCV(pipeline_local, grade_parametros, scoring='average_precision', cv=validacao_interna, n_jobs=-1, refit=True, error_score='raise')
            busca_grade.fit(quadro_modelagem_local.iloc[indices_treino][colunas], alvo[indices_treino], groups=grupos[indices_treino])
            predicoes[nome][indices_teste] = obter_probabilidade_classe_positiva(busca_grade.best_estimator_, quadro_modelagem_local.iloc[indices_teste][colunas], 1)
            row[f'{nome}_best_params'] = json.dumps(busca_grade.best_params_)
        id_particao[indices_teste] = f
        detalhes.append(row)
    if not all((np.isfinite(v).all() for v in predicoes.values())):
        raise RuntimeError('OOF incompleto.')
    return (quadro_modelagem_local, alvo, grupos, predicoes, id_particao, pd.DataFrame(detalhes))
QUADRO_MODELAGEM, ALVO, GRUPOS, PREDICOES, ID_PARTICAO, DETALHES_PARTICOES = validar_oof_aninhado_tres_modelos(DADOS_MODELO, MATRIZ_CONTAGENS_ASV)

def montar_linha_metricas(nome, p):
    return {'modelo': nome, 'n': len(ALVO), 'eventos': int(ALVO.sum()), 'prevalencia': float(ALVO.mean()), 'PR_AUC': float(average_precision_score(ALVO, p)), 'ROC_AUC': float(roc_auc_score(ALVO, p)), 'Brier': float(brier_score_loss(ALVO, p)), 'LogLoss': float(log_loss(ALVO, np.clip(p, 1e-06, 1 - 1e-06)))}
METRICAS_MODELOS_REAIS = pd.DataFrame([montar_linha_metricas('Clínico-espacial', PREDICOES['clinico']), montar_linha_metricas('Microbioma CLR aninhado', PREDICOES['microbioma']), montar_linha_metricas('Clínico + microbioma aninhado', PREDICOES['combinado'])])
display(METRICAS_MODELOS_REAIS)
display(DETALHES_PARTICOES)
````

````python
# 14. Reamostragem bootstrap por criança e valor incremental do microbioma
def executar_bootstrap_por_crianca(alvo, grupos, predicoes, n_bootstrap, seed):
    gerador_aleatorio = np.random.default_rng(seed)
    criancas = np.unique(grupos)
    linhas = []
    for b in range(int(n_bootstrap)):
        amostrados = gerador_aleatorio.choice(criancas, size=len(criancas), replace=True)
        indices = np.concatenate([np.flatnonzero(grupos == c) for c in amostrados])
        alvo_bootstrap = alvo[indices]
        if len(np.unique(alvo_bootstrap)) < 2:
            continue
        linhas.append({'bootstrap': b, 'delta_PR_AUC': average_precision_score(alvo_bootstrap, predicoes['combinado'][indices]) - average_precision_score(alvo_bootstrap, predicoes['clinico'][indices]), 'delta_ROC_AUC': roc_auc_score(alvo_bootstrap, predicoes['combinado'][indices]) - roc_auc_score(alvo_bootstrap, predicoes['clinico'][indices]), 'delta_Brier': brier_score_loss(alvo_bootstrap, predicoes['combinado'][indices]) - brier_score_loss(alvo_bootstrap, predicoes['clinico'][indices])})
    tabela_bootstrap = pd.DataFrame(linhas)
    resumo = []
    for c in ['delta_PR_AUC', 'delta_ROC_AUC', 'delta_Brier']:
        resumo.append({'metrica': c, 'media': tabela_bootstrap[c].mean(), 'mediana': tabela_bootstrap[c].median(), 'ic95_inferior': tabela_bootstrap[c].quantile(0.025), 'ic95_superior': tabela_bootstrap[c].quantile(0.975), 'P_delta_gt_0': (tabela_bootstrap[c] > 0).mean()})
    return (tabela_bootstrap, pd.DataFrame(resumo))
RESULTADOS_BOOTSTRAP, RESUMO_BOOTSTRAP = executar_bootstrap_por_crianca(ALVO, GRUPOS, PREDICOES, CONFIG.N_BOOTSTRAPS, CONFIG.SEMENTE)
display(RESUMO_BOOTSTRAP)
````

````python
# 15. Auditoria de desempenho abaixo do acaso e orientação da classe positiva
def auditar_desempenho_abaixo_acaso(alvo, predicoes, id_particao):
    linhas = []
    for nome, p in predicoes.items():
        roc_auc = roc_auc_score(alvo, p)
        pr_auc = average_precision_score(alvo, p)
        linhas.append({'modelo': nome, 'PR_AUC': pr_auc, 'ROC_AUC': roc_auc, 'prevalencia': float(alvo.mean()), 'pr_abaixo_prevalencia': pr_auc < alvo.mean(), 'roc_abaixo_0_5': roc_auc < 0.5, 'auc_invertida_apenas_diagnostico': 1 - roc_auc, 'auc_invertida_reportavel': False})
    return pd.DataFrame(linhas)
AUDITORIA_ABAIXO_ACASO = auditar_desempenho_abaixo_acaso(ALVO, PREDICOES, ID_PARTICAO)
display(AUDITORIA_ABAIXO_ACASO)
````

# 16. Braço visual real — aquisição, auditoria e treinamento

Esta versão usa **imagens reais** do benchmark público de cárie e executa
treinamento real do detector. O braço visual é metodologicamente separado do
Single-tooth ECC porque as coortes não são pareadas em nível de
`patient_id + visit_id + tooth_fdi`.

A sequência é:

`Zenodo → download + hash → extração → descoberta YOLO → auditoria de labels → treino → validação → métricas → pesos + manifesto`

### Regras científicas

- nenhuma imagem sintética é adicionada;
- nenhum label é fabricado;
- se a estrutura YOLO não puder ser inferida com segurança, a execução para;
- split por paciente só é declarado quando um identificador de paciente
  pode ser extraído por regra explicitamente validada;
- caso contrário, o notebook preserva o split fornecido pelo dataset e o
  relatório marca `patient_disjoint_verified = False`;
- detecção de cárie **não é automaticamente convertida em FDI** sem uma
  máscara/instância dentária compatível.

## 16.0 v1.9 — Adaptação universal de datasets e política de split

A v1.9 não pressupõe que um dataset Kaggle esteja organizado como
`images/ + labels/*.txt`. Antes da conversão, o notebook inventaria a estrutura
real e tenta adaptadores em ordem auditável:

1. YAML YOLO oficial;
2. TXT YOLO pareado globalmente por `stem`, independentemente da pasta;
3. LabelMe JSON;
4. COCO JSON;
5. Pascal VOC XML;
6. máscaras binárias/semânticas convertidas para bounding boxes;
7. CSV com bounding boxes em colunas reconhecíveis.

A decisão de split também é conservadora:

- split oficial do autor → preservar;
- `patient_id` confiável → split agrupado por paciente;
- sem split e sem `patient_id` → criar apenas `train/val` para desenvolvimento;
- **não fabricar um conjunto de teste supostamente independente em nível de imagem**;
- imagens sem label são registradas como `inference_only`.

Isto corrige diretamente o caso do **Children Caries Detection Dataset**, cuja
página informa imagens com rótulos de cárie, mas cuja organização interna não é
obrigada a seguir o layout YOLO convencional.

````python
# 16.1 Utilidades do braço visual real
from collections import Counter, defaultdict
EXTENSOES_IMAGEM = {'.jpg', '.jpeg', '.png', '.bmp', '.webp', '.tif', '.tiff'}

def auditar_gpu():
    informacoes = {'torch_available': False, 'cuda_available': False, 'device': 'cpu', 'gpu_name': None}
    try:
        import torch
        informacoes['torch_available'] = True
        informacoes['cuda_available'] = bool(torch.cuda.is_available())
        if torch.cuda.is_available():
            informacoes['device'] = 'cuda:0'
            informacoes['gpu_name'] = torch.cuda.get_device_name(0)
    except Exception:
        pass
    return informacoes

def extrair_zip_uma_vez(caminho_zip, destino):
    caminho_zip, destino = (Path(caminho_zip), Path(destino))
    destino.mkdir(parents=True, exist_ok=True)
    marcador = destino / '.extracted.ok'
    if marcador.exists():
        return destino
    with zipfile.ZipFile(caminho_zip) as quadro_modelagem_local:
        quadro_modelagem_local.extractall(destino)
    marcador.write_text(datetime.now(timezone.utc).isoformat())
    return destino

def eh_arquivo_rotulo_yolo(caminho):
    caminho = Path(caminho)
    if caminho.suffix.lower() != '.txt':
        return False
    try:
        linhas_texto = [x.strip() for x in caminho.read_text(errors='ignore').splitlines() if x.strip()]
    except Exception:
        return False
    if not linhas_texto:
        return True
    for linha_texto in linhas_texto[:50]:
        partes = linha_texto.split()
        if len(partes) != 5:
            return False
        try:
            id_classe = int(float(partes[0]))
            vals = [float(x) for x in partes[1:]]
        except Exception:
            return False
        if id_classe < 0 or any((v < 0 or v > 1 for v in vals)):
            return False
    return True

def descobrir_estrutura_yolo(raiz):
    raiz = Path(raiz)
    candidatos_yaml = list(raiz.rglob('data.yaml')) + list(raiz.rglob('dataset.yaml'))
    for caminho_yaml in candidatos_yaml:
        texto = caminho_yaml.read_text(errors='ignore')
        if 'train:' in texto and ('val:' in texto or 'test:' in texto):
            return {'mode': 'author_yaml', 'yaml': caminho_yaml, 'root': caminho_yaml.parent}
    diretorios = [p for p in raiz.rglob('*') if p.is_dir()]
    diretorios_imagens = [p for p in diretorios if p.name.lower() == 'images']
    candidatos = []
    for diretorio_imagens in diretorios_imagens:
        pasta_pai = diretorio_imagens.parent
        nome_particao = pasta_pai.name.lower()
        if nome_particao not in {'train', 'val', 'valid', 'validation', 'test'}:
            continue
        diretorio_rotulos = pasta_pai / 'labels'
        if diretorio_rotulos.exists():
            n_imagens = sum((p.suffix.lower() in EXTENSOES_IMAGEM for p in diretorio_imagens.rglob('*') if p.is_file()))
            n_rotulos = sum((p.suffix.lower() == '.txt' for p in diretorio_rotulos.rglob('*') if p.is_file()))
            candidatos.append((nome_particao, pasta_pai, diretorio_imagens, diretorio_rotulos, n_imagens, n_rotulos))
    if not candidatos:
        raise RuntimeError('Não foi possível identificar uma estrutura YOLO segura. Nenhum YAML do autor e nenhum layout split/images+labels convencional foi encontrado.')
    return {'mode': 'conventional_layout', 'splits': candidatos, 'root': raiz}

def construir_yaml_da_estrutura(estrutura, caminho_yaml_saida):
    import yaml
    por_particao = {}
    for nome_particao, pasta_pai, diretorio_imagens, diretorio_rotulos, n_imagens, n_rotulos in estrutura['splits']:
        nome_particao_canonica = {'valid': 'val', 'validation': 'val'}.get(nome_particao, nome_particao)
        por_particao[nome_particao_canonica] = diretorio_imagens
    if 'train' not in por_particao or 'val' not in por_particao:
        raise RuntimeError('Fallback YOLO exige pelo menos train e val explícitos.')
    ids_classes = set()
    for _, _, _, diretorio_rotulos, _, _ in estrutura['splits']:
        for caminho_rotulo in diretorio_rotulos.rglob('*.txt'):
            for linha_texto in caminho_rotulo.read_text(errors='ignore').splitlines():
                if not linha_texto.strip():
                    continue
                try:
                    ids_classes.add(int(float(linha_texto.split()[0])))
                except Exception:
                    pass
    if not ids_classes:
        raise RuntimeError('Nenhuma classe YOLO encontrada.')
    max_cls = max(ids_classes)
    nomes_classes = {i: 'caries' if i == 0 and len(ids_classes) == 1 else f'class_{i}' for i in range(max_cls + 1)}
    configuracao_yaml = {'path': str(estrutura['root'].resolve()), 'train': str(por_particao['train'].resolve()), 'val': str(por_particao['val'].resolve()), 'names': nomes_classes}
    if 'test' in por_particao:
        configuracao_yaml['test'] = str(por_particao['test'].resolve())
    caminho_yaml_saida = Path(caminho_yaml_saida)
    caminho_yaml_saida.write_text(yaml.safe_dump(configuracao_yaml, sort_keys=False, allow_unicode=True))
    return caminho_yaml_saida

def resolver_yaml_dataset(raiz_extraida):
    estrutura = descobrir_estrutura_yolo(raiz_extraida)
    if estrutura['mode'] == 'author_yaml':
        return (Path(estrutura['yaml']), estrutura)
    if importlib.util.find_spec('yaml') is None:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'pyyaml'], check=True)
    caminho_yaml_saida = PASTAS['imagens'] / 'generated_data.yaml'
    return (construir_yaml_da_estrutura(estrutura, caminho_yaml_saida), estrutura)

def resolver_caminhos_yolo(caminho_yaml_dataset):
    import yaml
    configuracao_yaml = yaml.safe_load(Path(caminho_yaml_dataset).read_text())
    pasta_base_yaml = Path(configuracao_yaml.get('path', Path(caminho_yaml_dataset).parent))
    if not pasta_base_yaml.is_absolute():
        pasta_base_yaml = (Path(caminho_yaml_dataset).parent / pasta_base_yaml).resolve()
    resolved = {}
    for chave in ['train', 'val', 'test']:
        if chave not in configuracao_yaml:
            continue
        val = configuracao_yaml[chave]
        vals = val if isinstance(val, list) else [val]
        caminhos = []
        for x in vals:
            p = Path(str(x))
            if not p.is_absolute():
                p = (pasta_base_yaml / p).resolve()
            caminhos.append(p)
        resolved[chave] = caminhos
    return (configuracao_yaml, resolved)

def iterar_imagens_entrada_dataset(entrada):
    p = Path(entrada)
    if p.is_dir():
        for x in p.rglob('*'):
            if x.is_file() and x.suffix.lower() in EXTENSOES_IMAGEM:
                yield x
    elif p.is_file() and p.suffix.lower() == '.txt':
        for linha_texto in p.read_text(errors='ignore').splitlines():
            q = Path(linha_texto.strip())
            if q.exists() and q.suffix.lower() in EXTENSOES_IMAGEM:
                yield q
    elif p.is_file() and p.suffix.lower() in EXTENSOES_IMAGEM:
        yield p

def obter_rotulo_candidato_imagem(imagem):
    partes = list(imagem.parts)
    indices_encontrados = [i for i, p in enumerate(partes) if p.lower() == 'images']
    if indices_encontrados:
        i = indices_encontrados[-1]
        caminho_rotulo = Path(*partes[:i], 'labels', *partes[i + 1:]).with_suffix('.txt')
        return caminho_rotulo
    return imagem.with_suffix('.txt')

def auditar_dataset_yolo(caminho_yaml_dataset):
    configuracao_yaml, caminhos_particoes = resolver_caminhos_yolo(caminho_yaml_dataset)
    linhas = []
    invalidos = []
    contagem_classes = Counter()
    grupos_pacientes = defaultdict(set)
    regex_paciente = re.compile(CONFIG.REGEX_ID_PACIENTE) if CONFIG.REGEX_ID_PACIENTE.strip() else None
    for particao_nome, entradas in caminhos_particoes.items():
        imagens = []
        for entrada in entradas:
            imagens.extend(list(iterar_imagens_entrada_dataset(entrada)))
        imagens = sorted(set(imagens))
        for imagem in imagens:
            rotulo = obter_rotulo_candidato_imagem(imagem)
            possui_rotulo = rotulo.exists()
            n_objetos = 0
            if possui_rotulo:
                linhas_texto = [x.strip() for x in rotulo.read_text(errors='ignore').splitlines() if x.strip()]
                for numero_linha, linha_texto in enumerate(linhas_texto, 1):
                    partes = linha_texto.split()
                    if len(partes) != 5:
                        invalidos.append((particao_nome, str(rotulo), numero_linha, 'field_count'))
                        continue
                    try:
                        id_classe = int(float(partes[0]))
                        vals = [float(x) for x in partes[1:]]
                    except Exception:
                        invalidos.append((particao_nome, str(rotulo), numero_linha, 'parse_error'))
                        continue
                    if id_classe < 0:
                        invalidos.append((particao_nome, str(rotulo), numero_linha, 'negative_class'))
                        continue
                    if any((v < 0 or v > 1 for v in vals)):
                        invalidos.append((particao_nome, str(rotulo), numero_linha, 'bbox_out_of_range'))
                        continue
                    if vals[2] <= 0 or vals[3] <= 0:
                        invalidos.append((particao_nome, str(rotulo), numero_linha, 'nonpositive_bbox'))
                        continue
                    contagem_classes[id_classe] += 1
                    n_objetos += 1
            if regex_paciente:
                correspondencia = regex_paciente.search(imagem.name)
                if correspondencia:
                    id_paciente_local = correspondencia.group(1) if correspondencia.groups() else correspondencia.group(0)
                    grupos_pacientes[id_paciente_local].add(particao_nome)
            linhas.append({'split': particao_nome, 'image': str(imagem), 'label': str(rotulo), 'rotulo_existe': possui_rotulo, 'n_objetos': n_objetos})
    arquivos = pd.DataFrame(linhas)
    tabela_invalidos = pd.DataFrame(invalidos, columns=['split', 'label', 'line', 'error'])
    sobreposicao_pacientes = {id_paciente_local: sorted(particoes) for id_paciente_local, particoes in grupos_pacientes.items() if len(particoes) > 1}
    identificacao_paciente_verificada = bool(regex_paciente) and len(grupos_pacientes) > 0
    pacientes_disjuntos = identificacao_paciente_verificada and len(sobreposicao_pacientes) == 0
    resumo = {'n_imagens': int(len(arquivos)), 'n_imagens_rotuladas': int(arquivos['rotulo_existe'].sum()) if len(arquivos) else 0, 'n_imagens_sem_rotulo': int((~arquivos['rotulo_existe']).sum()) if len(arquivos) else 0, 'n_objetos': int(arquivos['n_objetos'].sum()) if len(arquivos) else 0, 'n_linhas_rotulo_invalidas': int(len(tabela_invalidos)), 'classes': dict(contagem_classes), 'regex_id_paciente_configurada': bool(regex_paciente), 'n_grupos_pacientes_detectados': int(len(grupos_pacientes)), 'n_grupos_sobreposicao_pacientes': int(len(sobreposicao_pacientes)), 'pacientes_disjuntos_verificados': bool(pacientes_disjuntos)}
    return (resumo, arquivos, tabela_invalidos, sobreposicao_pacientes)
````

## 16.1.2 Curadoria de desempenho sem vazamento

Antes de qualquer tuning, cada dataset YOLO passa por uma camada de tratamento
que **não utiliza o conjunto de teste para tomar decisões**.

O tratamento executa:

1. inventário das imagens por split;
2. SHA-256 de cada imagem;
3. remoção de duplicatas exatas com prioridade `test > val > train`;
4. auditoria perceptual por dHash, sem remoção automática de near-duplicates;
5. distribuição de classes e áreas das caixas;
6. reamostragem limitada **somente do treino** quando há forte desbalanceamento;
7. geração de YAML curado, preservando `val` e `test`;
8. configuração de augmentation específica para cada domínio.

A finalidade é melhorar generalização e estabilidade sem contaminar a avaliação.

### Parâmetros de treinamento e justificativa

A implementação utiliza argumentos nativos do Ultralytics para `imgsz`, `batch`,
`patience`, `seed`, `deterministic`, `optimizer` e augmentations. A configuração
otimizada é deliberadamente conservadora em imagens médicas:

- panorâmicas: `mosaic=0`, `mixup=0`, `copy_paste=0`, `flipud=0` e HSV desativado;
- fotografias intraorais: jitter fotométrico leve e pequenas transformações;
- anatomia: augmentations moderadas, sem inversão vertical;
- `seed` e `deterministic=True` são mantidos para reprodutibilidade;
- resolução maior é testada para panorâmicas, nas quais lesões podem ocupar
  pequena área relativa.

O notebook compara uma configuração-base segura e uma configuração tratada
usando **somente o conjunto de validação**. O conjunto `test`, quando existente,
é reservado para a avaliação final do modelo selecionado.

````python
# 16.1.3 Funções de tratamento e tuning visual rigoroso
from collections import Counter, defaultdict
from PIL import Image

CONFIGURACOES_AUMENTACAO_DOMINIO = {
    "fotografia_intraoral": {
        "imgsz": lambda: CONFIG.TAMANHO_IMAGEM_INTRAORAL,
        "degrees": 5.0,
        "translate": 0.04,
        "scale": 0.12,
        "shear": 0.0,
        "perspective": 0.0,
        "flipud": 0.0,
        "fliplr": 0.5,
        "hsv_h": 0.010,
        "hsv_s": 0.20,
        "hsv_v": 0.20,
        "mosaic": 0.15,
        "mixup": 0.0,
        "copy_paste": 0.0,
        "close_mosaic": 5,
        "rect": False,
    },
    "panoramica_pediatrica": {
        "imgsz": lambda: CONFIG.TAMANHO_IMAGEM_PANORAMICA,
        "degrees": 2.0,
        "translate": 0.02,
        "scale": 0.08,
        "shear": 0.0,
        "perspective": 0.0,
        "flipud": 0.0,
        "fliplr": 0.5,
        "hsv_h": 0.0,
        "hsv_s": 0.0,
        "hsv_v": 0.0,
        "mosaic": 0.0,
        "mixup": 0.0,
        "copy_paste": 0.0,
        "close_mosaic": 0,
        "rect": True,
    },
    "anatomia_dentaria": {
        "imgsz": lambda: CONFIG.TAMANHO_IMAGEM_ANATOMIA,
        "degrees": 5.0,
        "translate": 0.05,
        "scale": 0.15,
        "shear": 0.0,
        "perspective": 0.0,
        "flipud": 0.0,
        "fliplr": 0.5,
        "hsv_h": 0.005,
        "hsv_s": 0.08,
        "hsv_v": 0.08,
        "mosaic": 0.20,
        "mixup": 0.0,
        "copy_paste": 0.0,
        "close_mosaic": 5,
        "rect": False,
    },
}

def calcular_dhash_imagem(caminho_imagem, tamanho=8):
    with Image.open(caminho_imagem) as imagem:
        cinza = imagem.convert("L").resize((tamanho + 1, tamanho))
        pixels = np.asarray(cinza, dtype=np.int16)
    diferencas = pixels[:, 1:] > pixels[:, :-1]
    valor = 0
    for bit in diferencas.flatten():
        valor = (valor << 1) | int(bit)
    return int(valor)

def distancia_hamming_inteiros(a, b):
    return int((int(a) ^ int(b)).bit_count())

class ArvoreBK:
    def __init__(self, distancia):
        self.distancia = distancia
        self.raiz = None

    def adicionar(self, valor, metadado):
        if self.raiz is None:
            self.raiz = [valor, metadado, {}]
            return
        no = self.raiz
        while True:
            d = self.distancia(valor, no[0])
            if d in no[2]:
                no = no[2][d]
            else:
                no[2][d] = [valor, metadado, {}]
                return

    def buscar(self, valor, raio):
        if self.raiz is None:
            return []
        encontrados = []
        pilha = [self.raiz]
        while pilha:
            no = pilha.pop()
            d = self.distancia(valor, no[0])
            if d <= raio:
                encontrados.append((d, no[1]))
            minimo, maximo = d - raio, d + raio
            for distancia_aresta, filho in no[2].items():
                if minimo <= distancia_aresta <= maximo:
                    pilha.append(filho)
        return encontrados

def listar_imagens_por_split(caminho_yaml_dataset):
    _, caminhos_particoes = resolver_caminhos_yolo(caminho_yaml_dataset)
    resultado = {}
    for split, entradas in caminhos_particoes.items():
        imagens = []
        for entrada in entradas:
            imagens.extend(iterar_imagens_entrada_dataset(entrada))
        resultado[split] = sorted(set(Path(x).resolve() for x in imagens))
    return resultado

def auditar_classes_e_caixas(caminho_yaml_dataset):
    imagens_splits = listar_imagens_por_split(caminho_yaml_dataset)
    contagens = Counter()
    linhas = []
    for split, imagens in imagens_splits.items():
        for imagem in imagens:
            rotulo = obter_rotulo_candidato_imagem(imagem)
            if not rotulo.exists():
                continue
            try:
                with Image.open(imagem) as im:
                    largura, altura = im.size
            except Exception:
                largura, altura = np.nan, np.nan
            for numero_linha, texto in enumerate(rotulo.read_text(errors="ignore").splitlines(), 1):
                if not texto.strip():
                    continue
                partes = texto.split()
                if len(partes) != 5:
                    continue
                try:
                    classe = int(float(partes[0]))
                    xc, yc, w, h = [float(x) for x in partes[1:]]
                except Exception:
                    continue
                contagens[(split, classe)] += 1
                linhas.append({
                    "split": split,
                    "classe": classe,
                    "imagem": str(imagem),
                    "linha": numero_linha,
                    "area_normalizada": float(w*h),
                    "largura_relativa": float(w),
                    "altura_relativa": float(h),
                    "area_pixels_aprox": float(w*h*largura*altura) if np.isfinite(largura) else np.nan,
                })
    caixas = pd.DataFrame(linhas)
    distribuicao = pd.DataFrame([
        {"split": split, "classe": classe, "n_objetos": n}
        for (split, classe), n in sorted(contagens.items())
    ])
    return distribuicao, caixas

def construir_listas_sem_duplicatas(caminho_yaml_dataset, pasta_saida, nome_dataset):
    pasta_saida = Path(pasta_saida)
    pasta_saida.mkdir(parents=True, exist_ok=True)
    imagens_splits = listar_imagens_por_split(caminho_yaml_dataset)

    prioridade = ["test", "val", "train"]
    hashes_vistos = {}
    retidas = {split: [] for split in imagens_splits}
    removidas = []

    for split in prioridade + [s for s in imagens_splits if s not in prioridade]:
        if split not in imagens_splits:
            continue
        for imagem in imagens_splits[split]:
            h = calcular_sha256(imagem)
            if h in hashes_vistos:
                removidas.append({
                    "imagem_removida": str(imagem),
                    "split_removido": split,
                    "imagem_mantida": hashes_vistos[h]["imagem"],
                    "split_mantido": hashes_vistos[h]["split"],
                    "sha256": h,
                })
                continue
            hashes_vistos[h] = {"imagem": str(imagem), "split": split}
            retidas[split].append(imagem)

    # near-duplicates: reportar, não remover automaticamente
    proximas = []
    if CONFIG.AUDITAR_DUPLICATAS_PERCEPTUAIS:
        arvore = ArvoreBK(distancia_hamming_inteiros)
        for split in ["test","val","train"]:
            for imagem in retidas.get(split, []):
                try:
                    dh = calcular_dhash_imagem(imagem)
                except Exception:
                    continue
                encontrados = arvore.buscar(dh, CONFIG.LIMIAR_HAMMING_DHASH)
                for distancia, meta in encontrados:
                    if meta["split"] != split:
                        proximas.append({
                            "imagem_a": meta["imagem"],
                            "split_a": meta["split"],
                            "imagem_b": str(imagem),
                            "split_b": split,
                            "distancia_hamming_dhash": distancia,
                        })
                arvore.adicionar(dh, {"imagem": str(imagem), "split": split})

    if "train" not in retidas or len(retidas["train"]) == 0:
        raise RuntimeError(f"{nome_dataset}: treino vazio após deduplicação.")
    if "val" not in retidas or len(retidas["val"]) == 0:
        raise RuntimeError(f"{nome_dataset}: validação vazia após deduplicação.")

    # listas absolutas curadas
    arquivos_listas = {}
    for split, imagens in retidas.items():
        lista = pasta_saida / f"{nome_dataset}_{split}.txt"
        lista.write_text("\n".join(str(p) for p in imagens) + ("\n" if imagens else ""))
        arquivos_listas[split] = lista

    return retidas, pd.DataFrame(removidas), pd.DataFrame(proximas), arquivos_listas

def calcular_fatores_balanceamento(caminho_yaml_dataset, max_fator=3):
    imagens_train = listar_imagens_por_split(caminho_yaml_dataset).get("train", [])
    classes_por_imagem = {}
    frequencia = Counter()
    for imagem in imagens_train:
        rotulo = obter_rotulo_candidato_imagem(imagem)
        classes = set()
        if rotulo.exists():
            for linha in rotulo.read_text(errors="ignore").splitlines():
                if not linha.strip():
                    continue
                try:
                    classes.add(int(float(linha.split()[0])))
                except Exception:
                    pass
        classes_por_imagem[imagem] = classes
        for classe in classes:
            frequencia[classe] += 1

    if len(frequencia) <= 1:
        return {imagem:1 for imagem in imagens_train}, dict(frequencia), False

    maior = max(frequencia.values())
    menor = max(1, min(frequencia.values()))
    razao = maior / menor
    if razao < CONFIG.LIMIAR_RAZAO_DESBALANCEAMENTO:
        return {imagem:1 for imagem in imagens_train}, dict(frequencia), False

    fatores = {}
    for imagem, classes in classes_por_imagem.items():
        fator = 1
        for classe in classes:
            alvo = int(np.ceil(np.sqrt(maior / max(1, frequencia[classe]))))
            fator = max(fator, min(int(max_fator), max(1, alvo)))
        fatores[imagem] = fator
    return fatores, dict(frequencia), True

def contar_duplicatas_exatas_entre_splits(caminho_yaml_dataset):
    imagens_splits = listar_imagens_por_split(caminho_yaml_dataset)
    por_hash = defaultdict(list)
    for split, imagens in imagens_splits.items():
        for imagem in imagens:
            por_hash[calcular_sha256(imagem)].append((split, str(imagem)))
    return sum(
        1 for itens in por_hash.values()
        if len({split for split,_ in itens}) > 1
    )

def tratar_dataset_yolo(caminho_yaml_dataset, dominio, nome_dataset):
    if dominio not in CONFIGURACOES_AUMENTACAO_DOMINIO:
        raise ValueError(f"Domínio desconhecido: {dominio}")

    import yaml
    configuracao_original, _ = resolver_caminhos_yolo(caminho_yaml_dataset)
    pasta_tratamento = PASTAS["imagens"] / "tratados" / nome_dataset
    pasta_tratamento.mkdir(parents=True, exist_ok=True)

    retidas, removidas, proximas, listas = construir_listas_sem_duplicatas(
        caminho_yaml_dataset, pasta_tratamento, nome_dataset
    )

    fatores, frequencia_train, balanceamento_aplicado = calcular_fatores_balanceamento(
        caminho_yaml_dataset,
        CONFIG.FATOR_MAXIMO_REAMOSTRAGEM,
    )

    # balanceamento somente do treino
    linhas_treino = []
    for imagem in retidas.get("train", []):
        fator = fatores.get(imagem, 1) if CONFIG.BALANCEAR_CLASSES_TREINO else 1
        linhas_treino.extend([str(imagem)] * int(fator))
    lista_treino = pasta_tratamento / f"{nome_dataset}_train_tratado.txt"
    lista_treino.write_text("\n".join(linhas_treino) + ("\n" if linhas_treino else ""))

    # val/test nunca recebem oversampling ou augmentation offline
    yaml_tratado = pasta_tratamento / f"{nome_dataset}_tratado.yaml"
    configuracao_nova = {
        "path": str(pasta_tratamento.resolve()),
        "train": str(lista_treino.resolve()),
        "val": str(listas["val"].resolve()) if "val" in listas else None,
        "names": configuracao_original.get("names", {}),
    }
    if "test" in listas and listas["test"].read_text().strip():
        configuracao_nova["test"] = str(listas["test"].resolve())
    configuracao_nova = {k:v for k,v in configuracao_nova.items() if v is not None}
    yaml_tratado.write_text(yaml.safe_dump(configuracao_nova, sort_keys=False, allow_unicode=True))

    distribuicao, caixas = auditar_classes_e_caixas(yaml_tratado)
    duplicatas_apos = contar_duplicatas_exatas_entre_splits(yaml_tratado)

    relatorio = {
        "dataset": nome_dataset,
        "dominio": dominio,
        "n_treino_original": len(listar_imagens_por_split(caminho_yaml_dataset).get("train", [])),
        "n_treino_unico": len(retidas.get("train", [])),
        "n_linhas_treino_apos_balanceamento": len(linhas_treino),
        "n_duplicatas_exatas_removidas": int(len(removidas)),
        "duplicatas_exatas_entre_splits": int(duplicatas_apos),
        "n_near_duplicates_reportados": int(len(proximas)),
        "balanceamento_aplicado": bool(balanceamento_aplicado and CONFIG.BALANCEAR_CLASSES_TREINO),
        "frequencia_classes_treino": frequencia_train,
        "augmentacao_apenas_treino": True,
        "test_usado_no_tratamento": False,
        "yaml_tratado": str(yaml_tratado),
    }

    removidas.to_csv(pasta_tratamento / "duplicatas_exatas_removidas.csv", index=False)
    proximas.to_csv(pasta_tratamento / "near_duplicates_reportados.csv", index=False)
    distribuicao.to_csv(pasta_tratamento / "distribuicao_classes.csv", index=False)
    caixas.to_csv(pasta_tratamento / "auditoria_caixas.csv", index=False)
    (pasta_tratamento / "relatorio_tratamento.json").write_text(
        json.dumps(relatorio, indent=2, ensure_ascii=False, default=str)
    )

    return yaml_tratado, relatorio, distribuicao, caixas

def argumentos_treino_por_dominio(dominio, otimizado=True):
    if not otimizado:
        return {
            "imgsz": 640,
            "optimizer": "auto",
            "degrees": 0.0,
            "translate": 0.0,
            "scale": 0.0,
            "shear": 0.0,
            "perspective": 0.0,
            "flipud": 0.0,
            "fliplr": 0.5,
            "hsv_h": 0.0 if dominio=="panoramica_pediatrica" else 0.005,
            "hsv_s": 0.0 if dominio=="panoramica_pediatrica" else 0.05,
            "hsv_v": 0.0 if dominio=="panoramica_pediatrica" else 0.05,
            "mosaic": 0.0,
            "mixup": 0.0,
            "copy_paste": 0.0,
            "close_mosaic": 0,
            "rect": dominio=="panoramica_pediatrica",
        }

    cfg = dict(CONFIGURACOES_AUMENTACAO_DOMINIO[dominio])
    cfg["imgsz"] = int(cfg["imgsz"]())
    cfg["optimizer"] = CONFIG.OTIMIZADOR_VISUAL
    cfg["lr0"] = CONFIG.TAXA_APRENDIZADO_VISUAL
    cfg["cos_lr"] = True
    return cfg

def extrair_metricas_ultralytics(metricas):
    d = getattr(metricas, "results_dict", {}) or {}
    return {
        "precision": float(d.get("metrics/precision(B)", np.nan)),
        "recall": float(d.get("metrics/recall(B)", np.nan)),
        "mAP50": float(d.get("metrics/mAP50(B)", np.nan)),
        "mAP50_95": float(d.get("metrics/mAP50-95(B)", np.nan)),
        "fitness": float(getattr(metricas, "fitness", np.nan)),
    }

def treinar_com_selecao_validacao(caminho_yaml_dataset, dominio, nome_execucao, epocas_finais=None):
    informacoes_gpu = auditar_gpu()
    if CONFIG.EXIGIR_GPU_TREINAMENTO_IMAGEM and not informacoes_gpu["cuda_available"]:
        raise RuntimeError("GPU CUDA não detectada. Ative GPU no Colab.")

    if importlib.util.find_spec("ultralytics") is None:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ultralytics"], check=True)
    from ultralytics import YOLO
    import yaml

    cfg_yaml = yaml.safe_load(Path(caminho_yaml_dataset).read_text())
    possui_test = bool(cfg_yaml.get("test"))
    epocas_finais = int(epocas_finais or CONFIG.EPOCAS_FINAIS_VISUAL)

    candidatos = [
        ("base_segura", argumentos_treino_por_dominio(dominio, otimizado=False)),
        ("tratado_otimizado", argumentos_treino_por_dominio(dominio, otimizado=True)),
    ]

    resultados_tuning = []
    if CONFIG.EXECUTAR_TUNING_VISUAL:
        for nome_candidato, args in candidatos:
            modelo = YOLO(CONFIG.PESOS_INICIAIS_MODELO_IMAGEM)
            resultado = modelo.train(
                data=str(caminho_yaml_dataset),
                epochs=CONFIG.EPOCAS_TUNING_VISUAL,
                batch=CONFIG.TAMANHO_LOTE_IMAGEM,
                patience=max(5, min(10, CONFIG.EPOCAS_TUNING_VISUAL)),
                workers=CONFIG.TRABALHADORES_CARREGAMENTO_IMAGEM,
                seed=CONFIG.SEMENTE,
                deterministic=True,
                cache=CONFIG.CACHE_TREINAMENTO_VISUAL,
                project=str(PASTAS["modelos"]),
                name=f"{nome_execucao}_tuning_{nome_candidato}",
                exist_ok=True,
                plots=False,
                verbose=True,
                **args,
            )
            melhor = Path(resultado.save_dir) / "weights" / "best.pt"
            avaliador = YOLO(str(melhor))
            metricas_val = avaliador.val(
                data=str(caminho_yaml_dataset),
                split="val",
                imgsz=int(args["imgsz"]),
                batch=CONFIG.TAMANHO_LOTE_IMAGEM,
                project=str(PASTAS["modelos"]),
                name=f"{nome_execucao}_tuning_{nome_candidato}_val",
                plots=False,
            )
            linha = {
                "modelo": nome_execucao,
                "dominio": dominio,
                "candidato": nome_candidato,
                "split_selecao": "val",
                **extrair_metricas_ultralytics(metricas_val),
                "argumentos": json.dumps(args, ensure_ascii=False, default=str),
            }
            resultados_tuning.append(linha)

        tabela_tuning = pd.DataFrame(resultados_tuning)
        if tabela_tuning["mAP50_95"].notna().any():
            melhor_nome = tabela_tuning.sort_values("mAP50_95", ascending=False).iloc[0]["candidato"]
        else:
            melhor_nome = "tratado_otimizado"
    else:
        tabela_tuning = pd.DataFrame()
        melhor_nome = "tratado_otimizado"

    argumentos_finais = dict(dict(candidatos)[melhor_nome])
    modelo_final = YOLO(CONFIG.PESOS_INICIAIS_MODELO_IMAGEM)
    treino_final = modelo_final.train(
        data=str(caminho_yaml_dataset),
        epochs=epocas_finais,
        batch=CONFIG.TAMANHO_LOTE_IMAGEM,
        patience=CONFIG.PACIENCIA_TREINAMENTO_IMAGEM,
        workers=CONFIG.TRABALHADORES_CARREGAMENTO_IMAGEM,
        seed=CONFIG.SEMENTE,
        deterministic=True,
        cache=CONFIG.CACHE_TREINAMENTO_VISUAL,
        project=str(PASTAS["modelos"]),
        name=f"{nome_execucao}_final_{melhor_nome}",
        exist_ok=True,
        plots=True,
        verbose=True,
        **argumentos_finais,
    )

    melhor_peso = Path(treino_final.save_dir) / "weights" / "best.pt"
    if not melhor_peso.exists():
        raise RuntimeError("best.pt não encontrado após treinamento final.")

    split_final = "test" if possui_test else "val"
    avaliador_final = YOLO(str(melhor_peso))
    metricas_finais_obj = avaliador_final.val(
        data=str(caminho_yaml_dataset),
        split=split_final,
        imgsz=int(argumentos_finais["imgsz"]),
        batch=CONFIG.TAMANHO_LOTE_IMAGEM,
        project=str(PASTAS["modelos"]),
        name=f"{nome_execucao}_avaliacao_final_{split_final}",
        plots=True,
    )

    metricas_finais = {
        "modelo": nome_execucao,
        "dominio": dominio,
        "configuracao_selecionada": melhor_nome,
        "conjunto_avaliacao_final": split_final,
        "test_independente_disponivel": bool(possui_test),
        **extrair_metricas_ultralytics(metricas_finais_obj),
        "best_pt": str(melhor_peso),
        "sha256_best_pt": calcular_sha256(melhor_peso),
        "diretorio_treino": str(treino_final.save_dir),
        "imgsz_final": int(argumentos_finais["imgsz"]),
    }
    return metricas_finais, tabela_tuning, treino_final, metricas_finais_obj

def criar_manifesto_reprodutibilidade():
    versoes = {
        "python": platform.python_version(),
        "numpy": getattr(np, "__version__", "NA"),
        "pandas": getattr(pd, "__version__", "NA"),
        "sklearn": getattr(sklearn, "__version__", "NA"),
    }
    return {
        "project": "OdontoCA",
        "version": "1.8.1",
        "real_data_only": bool(CONFIG.SOMENTE_DADOS_REAIS),
        "created_utc": datetime.now(timezone.utc).isoformat(),
        "software_versions": versoes,
        "sdd_tdd": {
            "sdd_version": "5.0",
            "tdd_version": "5.0",
            "n_requisitos": len(REQUISITOS_SDD),
            "n_testes_especificados": len(CASOS_TDD_ESPECIFICADOS),
        },
    }

# 16.1.4 AssociaÃ§Ã£o geomÃ©trica lesÃ£o-dente (IoL - Intersection over Lesion)
def calcular_iol_caixas(box_lesao, box_dente):
    """
    Calcula Intersection over Lesion (IoL): Ãrea(InterseÃ§Ã£o) / Ãrea(LesÃ£o).
    Garante atribuiÃ§Ã£o correta quando a cÃ¡rie Ã© substancialmente menor que o dente.
    """
    lx1, ly1, lx2, ly2 = box_lesao
    dx1, dy1, dx2, dy2 = box_dente

    inter_x1 = max(lx1, dx1)
    inter_y1 = max(ly1, dy1)
    inter_x2 = min(lx2, dx2)
    inter_y2 = min(ly2, dy2)

    if inter_x2 <= inter_x1 or inter_y2 <= inter_y1:
        return 0.0

    area_inter = (inter_x2 - inter_x1) * (inter_y2 - inter_y1)
    area_lesao = max(1e-6, (lx2 - lx1) * (ly2 - ly1))
    return float(area_inter / area_lesao)

def associar_lesoes_aos_dentes(dentes_detectados, lesoes_detectadas, limiar_iol=0.40):
    """
    Associa cada lesÃ£o detectada ao dente candidato com maior IoL.
    Retorna lista estruturada de lesÃµes com atribuiÃ§Ã£o de dente FDI e status de certeza.
    """
    associacoes = []
    for lesao in lesoes_detectadas:
        melhor_dente = None
        melhor_iol = 0.0
        for dente in dentes_detectados:
            iol = calcular_iol_caixas(lesao["bbox"], dente["bbox"])
            if iol > melhor_iol:
                melhor_iol = iol
                melhor_dente = dente

        dente_fdi = melhor_dente["dente_fdi"] if (melhor_dente and melhor_iol >= limiar_iol) else None
        associacoes.append({
            "id_lesao": lesao.get("id_lesao"),
            "bbox_lesao": lesao["bbox"],
            "conf_lesao": lesao.get("conf", 1.0),
            "dente_fdi_associado": dente_fdi,
            "iol": melhor_iol,
            "status": "ASSOCIADO" if dente_fdi is not None else "NAO_ASSOCIADO_AMBIGUO",
        })
    return associacoes
````

````python
# 16.2.1 Download, inventário e adaptação universal dos datasets Kaggle
# Guard de execução
_PRE_REQUISITOS_KAGGLE = [
    "CONFIG", "PASTAS", "criar_sessao_http", "calcular_sha256",
    "extrair_zip_uma_vez", "auditar_dataset_yolo", "resolver_yaml_dataset",
    "obter_rotulo_candidato_imagem", "eh_arquivo_rotulo_yolo",
    "EXTENSOES_IMAGEM", "tratar_dataset_yolo",
]
_faltantes_kaggle = [nome for nome in _PRE_REQUISITOS_KAGGLE if nome not in globals()]
if _faltantes_kaggle:
    raise RuntimeError(
        "Etapa Kaggle executada fora de ordem. "
        f"Pré-requisitos ausentes: {_faltantes_kaggle}."
    )

from requests.auth import HTTPBasicAuth
from PIL import Image, ImageStat, ImageFilter
import xml.etree.ElementTree as ET

REGISTRO_FALHAS_PARSING = []

def registrar_falha_parsing(etapa, arquivo, erro):
    REGISTRO_FALHAS_PARSING.append({
        "etapa": str(etapa),
        "arquivo": str(arquivo),
        "erro": f"{type(erro).__name__}: {erro}",
    })

def obter_credenciais_kaggle():
    usuario = os.environ.get("KAGGLE_USERNAME")
    chave = os.environ.get("KAGGLE_KEY")
    if usuario and chave:
        return usuario, chave, "variaveis_ambiente"

    arquivo = Path.home() / ".kaggle" / "kaggle.json"
    if arquivo.exists():
        try:
            dados = json.loads(arquivo.read_text())
            usuario = dados.get("username")
            chave = dados.get("key")
            if usuario and chave:
                return usuario, chave, "kaggle_json"
        except Exception as erro:
            registrar_falha_parsing("credenciais_kaggle", arquivo, erro)

    return None, None, "sem_credenciais"

def localizar_cache_legado(nome_local):
    candidatos = [
        Path("/content/OdontoCA_v1_8_SDD4_TDD4/00_brutos") / f"{nome_local}.zip",
        Path("/content/OdontoCA_v1_8_3/00_brutos") / f"{nome_local}.zip",
    ]
    for candidato in candidatos:
        if candidato.exists() and zipfile.is_zipfile(candidato):
            return candidato
    return None

def baixar_dataset_kaggle(slug, nome_local):
    destino = PASTAS["brutos"] / f"{nome_local}.zip"
    if destino.exists() and zipfile.is_zipfile(destino):
        print("Cache atual:", destino)
        return destino

    if CONFIG.REAPROVEITAR_CACHE_V18:
        legado = localizar_cache_legado(nome_local)
        if legado is not None:
            shutil.copy2(legado, destino)
            print("Reaproveitando cache legado:", legado)
            print("SHA-256:", calcular_sha256(destino))
            return destino

    url = f"https://www.kaggle.com/api/v1/datasets/download/{slug}"
    usuario, chave, origem_credencial = obter_credenciais_kaggle()
    autenticacao = HTTPBasicAuth(usuario, chave) if usuario and chave else None

    print(f"Baixando Kaggle: {slug}")
    print("Autenticação:", origem_credencial)

    resposta = criar_sessao_http().get(
        url, stream=True, timeout=1800, allow_redirects=True, auth=autenticacao
    )
    if resposta.status_code in {401, 403}:
        raise RuntimeError(
            f"Kaggle recusou {slug} ({resposta.status_code}). "
            "Use autenticação Kaggle segura."
        )
    resposta.raise_for_status()

    temporario = destino.with_suffix(".zip.part")
    with open(temporario, "wb") as arquivo_saida:
        for bloco in resposta.iter_content(1024 * 1024):
            if bloco:
                arquivo_saida.write(bloco)
    temporario.replace(destino)

    if not zipfile.is_zipfile(destino):
        raise RuntimeError(f"{slug}: download não resultou em ZIP válido.")

    print("ZIP:", destino)
    print("SHA-256:", calcular_sha256(destino))
    return destino

def extrair_dataset_kaggle(slug, nome_local):
    arquivo_zip = baixar_dataset_kaggle(slug, nome_local)
    destino = PASTAS["imagens"] / nome_local
    extrair_zip_uma_vez(arquivo_zip, destino)
    return destino, arquivo_zip

def normalizar_stem(nome):
    s = Path(nome).stem.lower().strip()
    sufixos = [
        "_label", "-label", "_labels", "-labels",
        "_annotation", "-annotation", "_annot", "-annot",
        "_mask", "-mask", "_masks", "-masks",
        "_caries", "-caries",
    ]
    alterou = True
    while alterou:
        alterou = False
        for sufixo in sufixos:
            if s.endswith(sufixo):
                s = s[:-len(sufixo)]
                alterou = True
    return re.sub(r"[^a-z0-9]+", "", s)

def inventariar_dataset_visual(raiz):
    raiz = Path(raiz)
    arquivos = [p for p in raiz.rglob("*") if p.is_file() and p.name != ".extracted.ok"]
    por_extensao = Counter(p.suffix.lower() or "<sem_ext>" for p in arquivos)

    imagens = [p for p in arquivos if p.suffix.lower() in EXTENSOES_IMAGEM]
    txt = [p for p in arquivos if p.suffix.lower() == ".txt"]
    jsons = [p for p in arquivos if p.suffix.lower() == ".json"]
    xmls = [p for p in arquivos if p.suffix.lower() == ".xml"]
    csvs = [p for p in arquivos if p.suffix.lower() == ".csv"]
    yamls = [p for p in arquivos if p.suffix.lower() in {".yaml", ".yml"}]

    diretorios = Counter(p.parent.name.lower() for p in arquivos)
    return {
        "raiz": str(raiz),
        "n_arquivos_total": len(arquivos),
        "n_imagens": len(imagens),
        "n_txt": len(txt),
        "n_json": len(jsons),
        "n_xml": len(xmls),
        "n_csv": len(csvs),
        "n_yaml": len(yamls),
        "extensoes": dict(por_extensao),
        "diretorios_mais_frequentes": dict(diretorios.most_common(20)),
    }

def exportar_inventario_dataset(raiz, nome_dataset):
    raiz = Path(raiz)
    linhas = []
    for p in sorted(x for x in raiz.rglob("*") if x.is_file() and x.name != ".extracted.ok"):
        linhas.append({
            "caminho": str(p),
            "relativo": str(p.relative_to(raiz)),
            "extensao": p.suffix.lower(),
            "tamanho_bytes": p.stat().st_size,
            "stem_normalizado": normalizar_stem(p.name),
            "split_inferido": nome_particao_por_caminho(p),
        })
    tabela = pd.DataFrame(linhas)
    destino = PASTAS["relatorios"] / f"inventario_{nome_dataset}.csv"
    tabela.to_csv(destino, index=False)
    return tabela, destino

def indexar_por_stem_global(raiz, extensoes=None):
    indice = defaultdict(list)
    for p in Path(raiz).rglob("*"):
        if not p.is_file():
            continue
        if extensoes is not None and p.suffix.lower() not in extensoes:
            continue
        indice[normalizar_stem(p.name)].append(p)
    return indice

def parear_imagens_txt_yolo_global(raiz):
    imagens = [
        p for p in Path(raiz).rglob("*")
        if p.is_file() and p.suffix.lower() in EXTENSOES_IMAGEM
    ]
    indice_txt = indexar_por_stem_global(raiz, {".txt"})
    pares = []
    nao_rotuladas = []

    for imagem in imagens:
        chave = normalizar_stem(imagem.name)
        candidatos = [
            p for p in indice_txt.get(chave, [])
            if eh_arquivo_rotulo_yolo(p)
        ]
        # Priorizar candidatos cujo caminho contém labels/label/annotation.
        candidatos = sorted(
            candidatos,
            key=lambda p: (
                0 if any(x in {q.lower() for q in p.parts}
                         for x in ["labels", "label", "annotations", "annotation"]) else 1,
                len(p.parts),
                str(p)
            )
        )
        if candidatos:
            pares.append((imagem, candidatos[0], nome_particao_por_caminho(imagem)))
        else:
            nao_rotuladas.append(imagem)

    return pares, nao_rotuladas

def localizar_imagem_para_json(caminho_json, dados_json):
    caminho_json = Path(caminho_json)
    indice_imagens = indexar_por_stem_global(caminho_json.parents[0], EXTENSOES_IMAGEM)

    caminho_indicado = dados_json.get("imagePath")
    if caminho_indicado:
        nome = Path(caminho_indicado).name
        candidatos = indice_imagens.get(normalizar_stem(nome), [])
        if candidatos:
            return candidatos[0]

    candidatos = indice_imagens.get(normalizar_stem(caminho_json.name), [])
    return candidatos[0] if candidatos else None

def nome_particao_por_caminho(caminho):
    componentes = [parte.lower() for parte in Path(caminho).parts]
    aliases = [
        ("train", "train"), ("training", "train"),
        ("val", "val"), ("valid", "val"), ("validation", "val"),
        ("test", "test"), ("testing", "test"),
    ]
    for chave, canonico in aliases:
        if chave in componentes:
            return canonico
    return None

def inferir_id_paciente_visual(caminho):
    if not CONFIG.REGEX_ID_PACIENTE.strip():
        return None
    m = re.search(CONFIG.REGEX_ID_PACIENTE, Path(caminho).name)
    if not m:
        return None
    return m.group(1) if m.groups() else m.group(0)

def dividir_registros_rotulados(registros, semente):
    """
    Política v1.9:
    - preserva splits oficiais;
    - cria val a partir de train se necessário;
    - nunca cria test artificial sem patient_id confiável.
    """
    registros = list(registros)
    rng = np.random.default_rng(semente)

    por_split = {"train": [], "val": [], "test": [], None: []}
    for i, r in enumerate(registros):
        por_split[r.get("particao_original")].append(i)

    patient_ids = [inferir_id_paciente_visual(r["imagem"]) for r in registros]
    patient_id_confiavel = (
        len(registros) > 0
        and all(x is not None for x in patient_ids)
        and len(set(patient_ids)) >= 3
    )

    # Registros sem split oficial
    indices_sem = por_split[None]
    if indices_sem:
        if patient_id_confiavel:
            grupos = defaultdict(list)
            for i in indices_sem:
                grupos[patient_ids[i]].append(i)
            ids = np.asarray(sorted(grupos))
            rng.shuffle(ids)
            n_val_grupos = max(1, int(round(CONFIG.FRACAO_VALIDACAO_SEM_SPLIT * len(ids))))
            ids_val = set(ids[:n_val_grupos].tolist())
            for pid, inds in grupos.items():
                por_split["val" if pid in ids_val else "train"].extend(inds)
        else:
            ordem = np.asarray(indices_sem)
            rng.shuffle(ordem)
            n_val = max(1, int(round(CONFIG.FRACAO_VALIDACAO_SEM_SPLIT * len(ordem))))
            por_split["val"].extend(ordem[:n_val].tolist())
            por_split["train"].extend(ordem[n_val:].tolist())

    # Se existe train oficial mas não val, separar val somente de train.
    if por_split["train"] and not por_split["val"]:
        if patient_id_confiavel:
            grupos_train = defaultdict(list)
            for i in por_split["train"]:
                grupos_train[patient_ids[i]].append(i)
            ids = np.asarray(sorted(grupos_train))
            rng.shuffle(ids)
            n_val_grupos = max(1, int(round(CONFIG.FRACAO_VALIDACAO_SEM_SPLIT * len(ids))))
            ids_val = set(ids[:n_val_grupos].tolist())
            novo_train, novo_val = [], []
            for pid, inds in grupos_train.items():
                (novo_val if pid in ids_val else novo_train).extend(inds)
            por_split["train"] = novo_train
            por_split["val"] = novo_val
        else:
            ordem = np.asarray(por_split["train"])
            rng.shuffle(ordem)
            n_val = max(1, int(round(CONFIG.FRACAO_VALIDACAO_SEM_SPLIT * len(ordem))))
            por_split["val"] = ordem[:n_val].tolist()
            por_split["train"] = ordem[n_val:].tolist()

    # Não criar test artificial na ausência de patient_id.
    teste_artificial_criado = False
    if not por_split["test"] and patient_id_confiavel and CONFIG.FRACAO_TESTE_SEM_SPLIT > 0:
        # Reservado para uso futuro; default v1.9 = 0.
        teste_artificial_criado = True

    if not por_split["train"] or not por_split["val"]:
        raise RuntimeError(
            f"Split insuficiente: train={len(por_split['train'])}, "
            f"val={len(por_split['val'])}."
        )

    return por_split, patient_id_confiavel, teste_artificial_criado

def gravar_dataset_yolo_de_pares(pares, nao_rotuladas, pasta_saida, nome_classe="caries"):
    pasta_saida = Path(pasta_saida)
    registros = [
        {"imagem": img, "rotulo": lab, "particao_original": split}
        for img, lab, split in pares
    ]
    particoes, patient_ok, teste_artificial = dividir_registros_rotulados(
        registros, CONFIG.SEMENTE
    )

    if pasta_saida.exists():
        shutil.rmtree(pasta_saida)
    for split in ["train", "val", "test", "inference_only"]:
        (pasta_saida / split / "images").mkdir(parents=True, exist_ok=True)
        if split != "inference_only":
            (pasta_saida / split / "labels").mkdir(parents=True, exist_ok=True)

    classes = set()
    auditoria = []
    for split in ["train", "val", "test"]:
        for seq, indice in enumerate(particoes[split]):
            r = registros[indice]
            imagem = Path(r["imagem"])
            rotulo = Path(r["rotulo"])
            nome = f"{seq:05d}_{imagem.name}"
            dst_img = pasta_saida / split / "images" / nome
            dst_lab = pasta_saida / split / "labels" / Path(nome).with_suffix(".txt").name
            shutil.copy2(imagem, dst_img)
            shutil.copy2(rotulo, dst_lab)
            for linha in dst_lab.read_text(errors="ignore").splitlines():
                if linha.strip():
                    classes.add(int(float(linha.split()[0])))
            auditoria.append({
                "split": split, "imagem": str(dst_img), "rotulo": str(dst_lab)
            })

    for seq, imagem in enumerate(nao_rotuladas):
        nome = f"{seq:05d}_{Path(imagem).name}"
        shutil.copy2(imagem, pasta_saida / "inference_only" / "images" / nome)

    if not classes:
        raise RuntimeError("Nenhuma classe YOLO encontrada nos pares globais.")

    nomes = {
        i: (nome_classe if len(classes) == 1 else f"classe_{i}")
        for i in range(max(classes)+1)
    }

    import yaml
    cfg = {
        "path": str(pasta_saida.resolve()),
        "train": "train/images",
        "val": "val/images",
        "names": nomes,
    }
    if particoes["test"]:
        cfg["test"] = "test/images"

    yaml_dataset = pasta_saida / "data.yaml"
    yaml_dataset.write_text(
        yaml.safe_dump(cfg, sort_keys=False, allow_unicode=True)
    )

    relatorio = {
        "formato_detectado": "yolo_txt_global",
        "n_pares_rotulados": len(pares),
        "n_inference_only": len(nao_rotuladas),
        "patient_id_confiavel": bool(patient_ok),
        "split_por_grupo": bool(patient_ok),
        "teste_artificial_criado": bool(teste_artificial),
        "inference_only_usado_como_ground_truth": False,
        "nivel_independencia": (
            "paciente" if patient_ok
            else "imagem_interna_sem_alegacao_paciente"
        ),
        "proveniencia_conversao": "TXT YOLO localizado por stem global; conteúdo preservado",
    }
    return yaml_dataset, pd.DataFrame(auditoria), relatorio

def converter_pascal_voc_para_yolo(raiz, pasta_saida):
    registros = []
    for xml in Path(raiz).rglob("*.xml"):
        try:
            tree = ET.parse(xml)
            root = tree.getroot()
            filename = root.findtext("filename")
            width = int(float(root.find("size/width").text))
            height = int(float(root.find("size/height").text))
            imagens = indexar_por_stem_global(raiz, EXTENSOES_IMAGEM).get(
                normalizar_stem(filename or xml.name), []
            )
            if not imagens:
                continue
            objetos = []
            for obj in root.findall("object"):
                nome = (obj.findtext("name") or "caries").strip()
                bb = obj.find("bndbox")
                x1 = float(bb.findtext("xmin"))
                y1 = float(bb.findtext("ymin"))
                x2 = float(bb.findtext("xmax"))
                y2 = float(bb.findtext("ymax"))
                objetos.append((nome, x1, y1, x2, y2))
            if objetos:
                registros.append({
                    "imagem": imagens[0], "largura": width, "altura": height,
                    "objetos": objetos,
                    "particao_original": nome_particao_por_caminho(xml)
                })
        except Exception as erro:
            registrar_falha_parsing("pascal_voc", xml, erro)
    return converter_registros_bbox_para_yolo(registros, pasta_saida, "pascal_voc_xml")

def converter_coco_para_yolo(raiz, pasta_saida):
    for arquivo_json in Path(raiz).rglob("*.json"):
        try:
            dados = json.loads(arquivo_json.read_text(errors="ignore"))
        except Exception as erro:
            registrar_falha_parsing("coco_json", arquivo_json, erro)
            continue
        if not (
            isinstance(dados, dict)
            and isinstance(dados.get("images"), list)
            and isinstance(dados.get("annotations"), list)
            and isinstance(dados.get("categories"), list)
        ):
            continue

        categorias = {int(c["id"]): str(c.get("name", c["id"])) for c in dados["categories"]}
        imgs = {int(x["id"]): x for x in dados["images"]}
        anns = defaultdict(list)
        for a in dados["annotations"]:
            anns[int(a["image_id"])].append(a)

        indice_img = indexar_por_stem_global(raiz, EXTENSOES_IMAGEM)
        registros = []
        for image_id, meta in imgs.items():
            candidatos = indice_img.get(normalizar_stem(meta.get("file_name","")), [])
            if not candidatos:
                continue
            objetos = []
            for a in anns.get(image_id, []):
                bbox = a.get("bbox")
                if not bbox or len(bbox) != 4:
                    continue
                x, y, w, h = map(float, bbox)
                if w <= 0 or h <= 0:
                    continue
                objetos.append((categorias.get(int(a["category_id"]), "caries"), x, y, x+w, y+h))
            if objetos:
                registros.append({
                    "imagem": candidatos[0],
                    "largura": int(meta["width"]),
                    "altura": int(meta["height"]),
                    "objetos": objetos,
                    "particao_original": nome_particao_por_caminho(candidatos[0]),
                })
        if registros:
            return converter_registros_bbox_para_yolo(registros, pasta_saida, "coco_json")
    raise RuntimeError("Nenhum COCO JSON compatível encontrado.")

def converter_labelme_para_yolo_v19(raiz, pasta_saida):
    indice_imagens = indexar_por_stem_global(raiz, EXTENSOES_IMAGEM)
    registros = []
    for arquivo_json in Path(raiz).rglob("*.json"):
        try:
            dados = json.loads(arquivo_json.read_text(errors="ignore"))
        except Exception as erro:
            registrar_falha_parsing("labelme_json", arquivo_json, erro)
            continue
        formas = dados.get("shapes")
        if not isinstance(formas, list):
            continue
        candidatos = []
        if dados.get("imagePath"):
            candidatos = indice_imagens.get(normalizar_stem(dados["imagePath"]), [])
        if not candidatos:
            candidatos = indice_imagens.get(normalizar_stem(arquivo_json.name), [])
        if not candidatos:
            continue
        imagem = candidatos[0]
        largura = dados.get("imageWidth")
        altura = dados.get("imageHeight")
        if not largura or not altura:
            try:
                with Image.open(imagem) as im:
                    largura, altura = im.size
            except Exception as erro:
                registrar_falha_parsing("labelme_dimensoes", imagem, erro)
                continue
        objetos = []
        for forma in formas:
            pontos = forma.get("points") or []
            if len(pontos) < 2:
                continue
            xs = [float(p[0]) for p in pontos]
            ys = [float(p[1]) for p in pontos]
            x1,x2 = min(xs),max(xs)
            y1,y2 = min(ys),max(ys)
            if x2 > x1 and y2 > y1:
                objetos.append((str(forma.get("label","caries")), x1,y1,x2,y2))
        if objetos:
            registros.append({
                "imagem": imagem, "largura": int(largura), "altura": int(altura),
                "objetos": objetos,
                "particao_original": nome_particao_por_caminho(arquivo_json),
            })
    return converter_registros_bbox_para_yolo(registros, pasta_saida, "labelme_json")

def converter_registros_bbox_para_yolo(registros, pasta_saida, formato):
    if not registros:
        raise RuntimeError(f"Nenhum registro bbox para formato {formato}.")
    classes = sorted({obj[0] for r in registros for obj in r["objetos"]})
    mapa = {nome:i for i,nome in enumerate(classes)}
    particoes, patient_ok, teste_artificial = dividir_registros_rotulados(registros, CONFIG.SEMENTE)

    pasta_saida = Path(pasta_saida)
    if pasta_saida.exists():
        shutil.rmtree(pasta_saida)
    for split in ["train","val","test"]:
        (pasta_saida/split/"images").mkdir(parents=True, exist_ok=True)
        (pasta_saida/split/"labels").mkdir(parents=True, exist_ok=True)

    for split in ["train","val","test"]:
        for seq, indice in enumerate(particoes[split]):
            r = registros[indice]
            img = Path(r["imagem"])
            nome = f"{seq:05d}_{img.name}"
            dst_img = pasta_saida/split/"images"/nome
            dst_lab = pasta_saida/split/"labels"/Path(nome).with_suffix(".txt").name
            shutil.copy2(img, dst_img)
            linhas = []
            for classe,x1,y1,x2,y2 in r["objetos"]:
                L,A = float(r["largura"]), float(r["altura"])
                x1,x2 = max(0,x1), min(L,x2)
                y1,y2 = max(0,y1), min(A,y2)
                if x2 <= x1 or y2 <= y1:
                    continue
                xc=((x1+x2)/2)/L; yc=((y1+y2)/2)/A
                w=(x2-x1)/L; h=(y2-y1)/A
                linhas.append(f"{mapa[classe]} {xc:.8f} {yc:.8f} {w:.8f} {h:.8f}")
            dst_lab.write_text("\n".join(linhas)+("\n" if linhas else ""))

    import yaml
    cfg = {
        "path": str(pasta_saida.resolve()),
        "train": "train/images", "val": "val/images",
        "names": {i:nome for nome,i in mapa.items()},
    }
    if particoes["test"]:
        cfg["test"] = "test/images"
    yaml_dataset = pasta_saida/"data.yaml"
    yaml_dataset.write_text(yaml.safe_dump(cfg, sort_keys=False, allow_unicode=True))

    relatorio = {
        "formato_detectado": formato,
        "n_pares_rotulados": len(registros),
        "n_inference_only": 0,
        "patient_id_confiavel": bool(patient_ok),
        "split_por_grupo": bool(patient_ok),
        "teste_artificial_criado": bool(teste_artificial),
        "inference_only_usado_como_ground_truth": False,
        "nivel_independencia": "paciente" if patient_ok else "imagem_interna_sem_alegacao_paciente",
        "proveniencia_conversao": f"{formato} → bounding boxes YOLO normalizadas",
    }
    return yaml_dataset, pd.DataFrame(), relatorio

def converter_mascaras_para_yolo(raiz, pasta_saida):
    imagens = [p for p in Path(raiz).rglob("*") if p.is_file() and p.suffix.lower() in EXTENSOES_IMAGEM]
    masks = [
        p for p in imagens
        if any(token in {x.lower() for x in p.parts} for token in ["mask","masks","label","labels"])
        or "mask" in p.stem.lower()
    ]
    imagens_clinicas = [p for p in imagens if p not in masks]
    indice_img = defaultdict(list)
    for p in imagens_clinicas:
        indice_img[normalizar_stem(p.name)].append(p)

    registros = []
    for mascara in masks:
        candidatos = indice_img.get(normalizar_stem(mascara.name), [])
        if not candidatos:
            continue
        imagem = candidatos[0]
        try:
            arr = np.asarray(Image.open(mascara).convert("L"))
            ys, xs = np.where(arr > 0)
            if len(xs) == 0:
                continue
            # Uma máscara pode conter várias lesões; componentes conectados quando scipy disponível.
            try:
                from scipy import ndimage
                comp, ncomp = ndimage.label(arr > 0)
                objetos = []
                for k in range(1, ncomp+1):
                    yy, xx = np.where(comp == k)
                    if len(xx) < 5:
                        continue
                    objetos.append(("caries", float(xx.min()), float(yy.min()),
                                    float(xx.max()+1), float(yy.max()+1)))
            except Exception as erro:
                registrar_falha_parsing("componentes_mascara", mascara, erro)
                objetos = [("caries", float(xs.min()), float(ys.min()),
                            float(xs.max()+1), float(ys.max()+1))]
            with Image.open(imagem) as im:
                largura, altura = im.size
            registros.append({
                "imagem": imagem, "largura": largura, "altura": altura,
                "objetos": objetos,
                "particao_original": nome_particao_por_caminho(imagem),
            })
        except Exception as erro:
            registrar_falha_parsing("mascara_binaria", mascara, erro)

    return converter_registros_bbox_para_yolo(registros, pasta_saida, "mascara_binaria")

def converter_csv_bbox_para_yolo(raiz, pasta_saida):
    aliases = {
        "arquivo": ["filename","file","image","image_name","img","path"],
        "x1": ["xmin","x_min","x1","left"],
        "y1": ["ymin","y_min","y1","top"],
        "x2": ["xmax","x_max","x2","right"],
        "y2": ["ymax","y_max","y2","bottom"],
        "classe": ["class","label","category","name"],
    }
    indice_img = indexar_por_stem_global(raiz, EXTENSOES_IMAGEM)
    for csv in Path(raiz).rglob("*.csv"):
        try:
            df = pd.read_csv(csv)
        except Exception as erro:
            registrar_falha_parsing("csv_bbox", csv, erro)
            continue
        mapa_col = {c.lower():c for c in df.columns}
        escolhido = {}
        for chave, candidatos in aliases.items():
            for cand in candidatos:
                if cand in mapa_col:
                    escolhido[chave] = mapa_col[cand]
                    break
        if not {"arquivo","x1","y1","x2","y2"}.issubset(escolhido):
            continue

        registros_dict = {}
        for _, row in df.iterrows():
            candidatos = indice_img.get(normalizar_stem(str(row[escolhido["arquivo"]])), [])
            if not candidatos:
                continue
            img = candidatos[0]
            try:
                with Image.open(img) as im:
                    largura, altura = im.size
                obj = (
                    str(row[escolhido["classe"]]) if "classe" in escolhido else "caries",
                    float(row[escolhido["x1"]]), float(row[escolhido["y1"]]),
                    float(row[escolhido["x2"]]), float(row[escolhido["y2"]]),
                )
            except Exception as erro:
                registrar_falha_parsing("csv_bbox_linha", csv, erro)
                continue
            chave = str(img)
            registros_dict.setdefault(chave, {
                "imagem": img, "largura": largura, "altura": altura,
                "objetos": [], "particao_original": nome_particao_por_caminho(img),
            })["objetos"].append(obj)
        if registros_dict:
            return converter_registros_bbox_para_yolo(
                list(registros_dict.values()), pasta_saida, "csv_bbox"
            )
    raise RuntimeError("Nenhum CSV bbox compatível.")

def auditar_qualidade_imagens_yolo(caminho_yaml_dataset, nome_dataset):
    imagens_splits = listar_imagens_por_split(caminho_yaml_dataset)
    linhas = []
    for split, imagens in imagens_splits.items():
        for caminho in imagens:
            try:
                with Image.open(caminho) as im:
                    cinza = np.asarray(im.convert("L"), dtype=np.float32)
                    largura, altura = im.size
                media = float(cinza.mean())
                contraste = float(cinza.std())
                try:
                    from scipy import ndimage
                    lap = ndimage.laplace(cinza)
                    lap_var = float(lap.var())
                except Exception as erro:
                    registrar_falha_parsing("qc_laplaciano", caminho, erro)
                    lap_var = np.nan
                linhas.append({
                    "dataset": nome_dataset,
                    "split": split,
                    "imagem": str(caminho),
                    "largura": int(largura),
                    "altura": int(altura),
                    "media_intensidade": media,
                    "contraste": contraste,
                    "laplaciano_var": lap_var,
                    "blur_suspeito": bool(np.isfinite(lap_var) and lap_var < CONFIG.LIMIAR_BLUR_LAPLACIANO),
                    "baixo_contraste": bool(contraste < CONFIG.LIMIAR_BAIXO_CONTRASTE),
                    "subexposta": bool(media < CONFIG.LIMIAR_SUBEXPOSICAO_MEDIA),
                    "superexposta": bool(media > CONFIG.LIMIAR_SUPEREXPOSICAO_MEDIA),
                })
            except Exception as erro:
                registrar_falha_parsing("qc_imagem", caminho, erro)
    tabela = pd.DataFrame(linhas)
    tabela.to_csv(PASTAS["relatorios"]/f"qc_{nome_dataset}.csv", index=False)
    return tabela

def preparar_dataset_anotacoes_auto(raiz, nome_saida, classe_padrao="caries"):
    raiz = Path(raiz)
    inventario_resumo = inventariar_dataset_visual(raiz)
    tabela_inventario, arquivo_inventario = exportar_inventario_dataset(raiz, nome_saida)
    print("Inventário:", inventario_resumo)

    # 1) YAML autor
    try:
        resultado = resolver_yaml_dataset(raiz)
        yaml_dataset, estrutura = resultado
        resumo, arquivos, invalidos, _ = auditar_dataset_yolo(yaml_dataset)
        if resumo["n_imagens_rotuladas"] >= CONFIG.MIN_PARES_ROTULADOS_VISUAL:
            rel = {
                "formato_detectado": "yolo_autor",
                "n_pares_rotulados": resumo["n_imagens_rotuladas"],
                "n_inference_only": resumo["n_imagens_sem_rotulo"],
                "patient_id_confiavel": resumo["pacientes_disjuntos_verificados"],
                "split_por_grupo": resumo["pacientes_disjuntos_verificados"],
                "teste_artificial_criado": False,
                "inference_only_usado_como_ground_truth": False,
                "nivel_independencia": (
                    "paciente" if resumo["pacientes_disjuntos_verificados"]
                    else "split_autor_sem_alegacao_paciente"
                ),
                "proveniencia_conversao": "YAML oficial do dataset preservado",
                "arquivo_inventario": str(arquivo_inventario),
            }
            return yaml_dataset, estrutura, rel
    except Exception as erro:
        registrar_falha_parsing("detectar_yolo_autor", raiz, erro)

    # 2) TXT YOLO global por stem
    pares, nao_rotuladas = parear_imagens_txt_yolo_global(raiz)
    if len(pares) >= CONFIG.MIN_PARES_ROTULADOS_VISUAL:
        yaml_dataset, auditoria, rel = gravar_dataset_yolo_de_pares(
            pares, nao_rotuladas,
            PASTAS["imagens"]/f"{nome_saida}_yolo_global",
            classe_padrao,
        )
        rel["arquivo_inventario"] = str(arquivo_inventario)
        return yaml_dataset, {"mode":"yolo_txt_global"}, rel

    # 3) LabelMe
    if inventario_resumo["n_json"] > 0:
        try:
            yaml_dataset, _, rel = converter_labelme_para_yolo_v19(
                raiz, PASTAS["imagens"]/f"{nome_saida}_labelme_yolo"
            )
            rel["arquivo_inventario"] = str(arquivo_inventario)
            return yaml_dataset, {"mode":"labelme_json"}, rel
        except Exception as erro:
            registrar_falha_parsing("adaptador_labelme", raiz, erro)

        # 4) COCO
        try:
            yaml_dataset, _, rel = converter_coco_para_yolo(
                raiz, PASTAS["imagens"]/f"{nome_saida}_coco_yolo"
            )
            rel["arquivo_inventario"] = str(arquivo_inventario)
            return yaml_dataset, {"mode":"coco_json"}, rel
        except Exception as erro:
            registrar_falha_parsing("adaptador_coco", raiz, erro)

    # 5) Pascal VOC
    if inventario_resumo["n_xml"] > 0:
        try:
            yaml_dataset, _, rel = converter_pascal_voc_para_yolo(
                raiz, PASTAS["imagens"]/f"{nome_saida}_voc_yolo"
            )
            rel["arquivo_inventario"] = str(arquivo_inventario)
            return yaml_dataset, {"mode":"pascal_voc_xml"}, rel
        except Exception as erro:
            registrar_falha_parsing("adaptador_voc", raiz, erro)

    # 6) CSV bbox
    if inventario_resumo["n_csv"] > 0:
        try:
            yaml_dataset, _, rel = converter_csv_bbox_para_yolo(
                raiz, PASTAS["imagens"]/f"{nome_saida}_csv_yolo"
            )
            rel["arquivo_inventario"] = str(arquivo_inventario)
            return yaml_dataset, {"mode":"csv_bbox"}, rel
        except Exception as erro:
            registrar_falha_parsing("adaptador_csv", raiz, erro)

    # 7) Máscaras
    try:
        yaml_dataset, _, rel = converter_mascaras_para_yolo(
            raiz, PASTAS["imagens"]/f"{nome_saida}_mask_yolo"
        )
        rel["arquivo_inventario"] = str(arquivo_inventario)
        return yaml_dataset, {"mode":"mascara_binaria"}, rel
    except Exception as erro:
        registrar_falha_parsing("adaptador_mascara", raiz, erro)

    pd.DataFrame(REGISTRO_FALHAS_PARSING).to_csv(
        PASTAS["relatorios"]/f"falhas_parsing_{nome_saida}.csv", index=False
    )
    raise RuntimeError(
        f"{nome_saida}: nenhum formato de anotação suportado foi reconhecido. "
        f"Inventário salvo em {arquivo_inventario}. "
        f"Resumo={inventario_resumo}"
    )

# ------------------------------
# Downloads reais
# ------------------------------
RAIZ_KAGGLE_ANATOMIA, ZIP_KAGGLE_ANATOMIA = extrair_dataset_kaggle(
    CONFIG.SLUG_KAGGLE_ANATOMIA, "kaggle_anatomia_yolov8"
)
RAIZ_KAGGLE_CARIE_INFANTIL, ZIP_KAGGLE_CARIE_INFANTIL = extrair_dataset_kaggle(
    CONFIG.SLUG_KAGGLE_CARIE_INFANTIL, "kaggle_carie_infantil"
)
RAIZ_KAGGLE_PANORAMICO, ZIP_KAGGLE_PANORAMICO = extrair_dataset_kaggle(
    CONFIG.SLUG_KAGGLE_PANORAMICO_PEDIATRICO, "kaggle_panorama_pediatrico"
)

# ------------------------------
# Anatomia
# ------------------------------
YAML_KAGGLE_ANATOMIA, ESTRUTURA_KAGGLE_ANATOMIA, RELATORIO_PREPARACAO_ANATOMIA = (
    preparar_dataset_anotacoes_auto(
        RAIZ_KAGGLE_ANATOMIA, "anatomia_kaggle", "anatomia"
    )
)
AUDITORIA_KAGGLE_ANATOMIA, _, INVALIDOS_KAGGLE_ANATOMIA, _ = auditar_dataset_yolo(
    YAML_KAGGLE_ANATOMIA
)

# ------------------------------
# Children Caries Detection
# CORREÇÃO DO ERRO "0 pares imagem/label YOLO"
# ------------------------------
YAML_KAGGLE_CARIE_INFANTIL, ESTRUTURA_KAGGLE_CARIE, RELATORIO_PREPARACAO_CARIE = (
    preparar_dataset_anotacoes_auto(
        RAIZ_KAGGLE_CARIE_INFANTIL, "carie_infantil_kaggle", "caries"
    )
)
MODO_PREPARACAO_CARIE = RELATORIO_PREPARACAO_CARIE["formato_detectado"]
AUDITORIA_KAGGLE_CARIE, _, INVALIDOS_KAGGLE_CARIE, _ = auditar_dataset_yolo(
    YAML_KAGGLE_CARIE_INFANTIL
)

if AUDITORIA_KAGGLE_CARIE["n_imagens_rotuladas"] < CONFIG.MIN_PARES_ROTULADOS_VISUAL:
    raise RuntimeError(
        "Children Caries Detection Dataset foi adaptado, mas não possui "
        "pares rotulados suficientes para treinamento."
    )

# ------------------------------
# Zhang et al. — usar adaptador universal; preserva JSON LabelMe quando presente
# ------------------------------
YAML_KAGGLE_DOENCAS, ESTRUTURA_KAGGLE_DOENCAS, RELATORIO_PREPARACAO_DOENCAS = (
    preparar_dataset_anotacoes_auto(
        RAIZ_KAGGLE_PANORAMICO, "zhang_doencas", "caries"
    )
)
AUDITORIA_KAGGLE_DOENCAS, _, INVALIDOS_KAGGLE_DOENCAS, _ = auditar_dataset_yolo(
    YAML_KAGGLE_DOENCAS
)
CLASSES_ZHANG = sorted(AUDITORIA_KAGGLE_DOENCAS["classes"])

# ------------------------------
# Quality Control
# ------------------------------
AUDITORIA_QUALIDADE_ANATOMIA = auditar_qualidade_imagens_yolo(
    YAML_KAGGLE_ANATOMIA, "kaggle_anatomia"
)
AUDITORIA_QUALIDADE_CARIE = auditar_qualidade_imagens_yolo(
    YAML_KAGGLE_CARIE_INFANTIL, "kaggle_carie_pediatrica"
)
AUDITORIA_QUALIDADE_DOENCAS = auditar_qualidade_imagens_yolo(
    YAML_KAGGLE_DOENCAS, "kaggle_doencas_pediatricas"
)

# ------------------------------
# Resumo antes do tratamento
# ------------------------------
RESUMO_DATASETS_KAGGLE = pd.DataFrame([
    {
        "dataset": "Dental Anatomy Dataset - YOLOv8",
        "modo_preparacao": RELATORIO_PREPARACAO_ANATOMIA["formato_detectado"],
        **AUDITORIA_KAGGLE_ANATOMIA,
    },
    {
        "dataset": "Children Caries Detection Dataset",
        "modo_preparacao": RELATORIO_PREPARACAO_CARIE["formato_detectado"],
        **AUDITORIA_KAGGLE_CARIE,
    },
    {
        "dataset": "Children's Dental Panoramic Radiographs",
        "modo_preparacao": RELATORIO_PREPARACAO_DOENCAS["formato_detectado"],
        **AUDITORIA_KAGGLE_DOENCAS,
    },
])
display(RESUMO_DATASETS_KAGGLE)
display(pd.DataFrame([
    RELATORIO_PREPARACAO_ANATOMIA,
    RELATORIO_PREPARACAO_CARIE,
    RELATORIO_PREPARACAO_DOENCAS,
]))

# ------------------------------
# Tratamento anti-leakage existente
# ------------------------------
(
    YAML_KAGGLE_ANATOMIA_TRATADO,
    RELATORIO_TRATAMENTO_ANATOMIA,
    DISTRIBUICAO_CLASSES_ANATOMIA,
    CAIXAS_ANATOMIA,
) = tratar_dataset_yolo(
    YAML_KAGGLE_ANATOMIA, "anatomia_dentaria", "kaggle_anatomia"
)

(
    YAML_KAGGLE_CARIE_TRATADO,
    RELATORIO_TRATAMENTO_CARIE,
    DISTRIBUICAO_CLASSES_CARIE,
    CAIXAS_CARIE,
) = tratar_dataset_yolo(
    YAML_KAGGLE_CARIE_INFANTIL, "panoramica_pediatrica", "kaggle_carie_pediatrica"
)

(
    YAML_KAGGLE_DOENCAS_TRATADO,
    RELATORIO_TRATAMENTO_DOENCAS,
    DISTRIBUICAO_CLASSES_DOENCAS,
    CAIXAS_DOENCAS,
) = tratar_dataset_yolo(
    YAML_KAGGLE_DOENCAS, "panoramica_pediatrica", "kaggle_doencas_pediatricas"
)

AUDITORIA_KAGGLE_ANATOMIA_TRATADA, _, INVALIDOS_ANATOMIA_TRATADOS, _ = auditar_dataset_yolo(
    YAML_KAGGLE_ANATOMIA_TRATADO
)
AUDITORIA_KAGGLE_CARIE_TRATADA, _, INVALIDOS_CARIE_TRATADOS, _ = auditar_dataset_yolo(
    YAML_KAGGLE_CARIE_TRATADO
)
AUDITORIA_KAGGLE_DOENCAS_TRATADA, _, INVALIDOS_DOENCAS_TRATADOS, _ = auditar_dataset_yolo(
    YAML_KAGGLE_DOENCAS_TRATADO
)

pd.DataFrame(REGISTRO_FALHAS_PARSING).to_csv(
    PASTAS["relatorios"]/"registro_falhas_parsing_visual.csv", index=False
)

display(pd.DataFrame([
    RELATORIO_TRATAMENTO_ANATOMIA,
    RELATORIO_TRATAMENTO_CARIE,
    RELATORIO_TRATAMENTO_DOENCAS,
]))
````

````python
# 18. Regra de segurança para fusão multimodal entre coortes
class ErroFusaoEntreCoortes(RuntimeError):
    pass

def validar_fusao_multimodal(fonte_visual, fonte_longitudinal, mesmas_chaves_paciente=False):
    if fonte_visual != fonte_longitudinal or not mesmas_chaves_paciente:
        raise ErroFusaoEntreCoortes('Fusão multimodal em nível de paciente bloqueada: coortes não pareadas.')
    return True
try:
    validar_fusao_multimodal('zenodo-caries', 'single-tooth-ecc', False)
    raise RuntimeError('Guardrail de fusão falhou.')
except ErroFusaoEntreCoortes:
    print('CRITÉRIO DE FUSÃO ENTRE COORTES: APROVADO (fusão bloqueada)')
````

````python
# 19. Interoperabilidade experimental — HL7 FHIR
def classificar_faixa_risco(p):
    p = float(p)
    return 'low' if p < 0.3 else 'moderate' if p < 0.7 else 'high'

def criar_avaliacao_risco_fhir(id_paciente, dente_fdi, probabilidade):
    p = float(probabilidade)
    if not 0 <= p <= 1:
        raise ValueError('Probabilidade inválida.')
    return {'resourceType': 'RiskAssessment', 'status': 'preliminary', 'subject': {'reference': f'Patient/{id_paciente}'}, 'occurrenceDateTime': datetime.now(timezone.utc).isoformat(), 'prediction': [{'outcome': {'text': f'Experimental tooth-level caries progression risk: {dente_fdi}'}, 'probabilityDecimal': round(p, 6), 'qualitativeRisk': {'text': classificar_faixa_risco(p)}}], 'note': [{'text': 'Research output only. Not clinically validated.'}]}
````

# 19.1 Execução da suíte TDD 5.0 e rastreabilidade SDD → TDD

`SKIP` é permitido apenas quando a condição depende de informação que o dataset
não fornece de forma confiável, como identificador de paciente nas imagens.
`FAIL` bloqueia a promoção do pipeline.

````python
# 19.1.1 Runner TDD
def executar_suite_tdd():
    resultados = []
    for caso in CASOS_TDD_ESPECIFICADOS:
        try:
            caso["funcao"]()
            status = "PASS"
            detalhe = ""
        except TesteIgnorado as erro:
            status = "SKIP"
            detalhe = str(erro)
        except Exception as erro:
            status = "FAIL"
            detalhe = f"{type(erro).__name__}: {erro}"
        resultados.append({
            "id_teste": caso["id_teste"],
            "nivel": caso["nivel"],
            "requisitos": ";".join(caso["requisitos"]),
            "descricao": caso["descricao"],
            "status": status,
            "detalhe": detalhe,
        })
    return pd.DataFrame(resultados)

RELATORIO_TDD = executar_suite_tdd()
display(RELATORIO_TDD)

RESUMO_TDD = (
    RELATORIO_TDD["status"]
    .value_counts()
    .reindex(["PASS","SKIP","FAIL"], fill_value=0)
    .rename_axis("status")
    .reset_index(name="n")
)
display(RESUMO_TDD)

# rastreabilidade
linhas_rastreabilidade = []
for caso in CASOS_TDD_ESPECIFICADOS:
    status = RELATORIO_TDD.loc[
        RELATORIO_TDD["id_teste"].eq(caso["id_teste"]), "status"
    ].iloc[0]
    for requisito in caso["requisitos"]:
        linhas_rastreabilidade.append({
            "requisito_sdd": requisito,
            "teste_tdd": caso["id_teste"],
            "nivel": caso["nivel"],
            "status_teste": status,
        })

RASTREABILIDADE_SDD_TDD = pd.DataFrame(linhas_rastreabilidade)

ids_obrigatorios = {
    r["id"] for r in REQUISITOS_SDD if r["obrigatorio"]
}
ids_cobertos = set(RASTREABILIDADE_SDD_TDD["requisito_sdd"])
COBERTURA_SDD = len(ids_obrigatorios & ids_cobertos) / len(ids_obrigatorios)

print(f"Cobertura SDD→TDD: {COBERTURA_SDD:.1%}")
print("FAIL TDD:", int((RELATORIO_TDD["status"]=="FAIL").sum()))

if (RELATORIO_TDD["status"]=="FAIL").any():
    display(RELATORIO_TDD.loc[RELATORIO_TDD["status"]=="FAIL"])
    raise RuntimeError("TDD 5.0 BLOQUEOU O PIPELINE: existe pelo menos um FAIL.")
````
